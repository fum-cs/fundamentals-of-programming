## الف) جعبهٔ سیاه و تعریف تابع

**تمرین 1** · *Explain Why* · ★ (پایه)

دو جعبهٔ زیر را در نظر بگیرید: جعبهٔ ۱ هر عدد را در ۲ ضرب می‌کند و جعبهٔ ۲ هر عدد را با خودش جمع می‌کند. آیا این دو «یک تابع» هستند یا «دو تابع»؟ آیا «یک الگوریتم» هستند یا دو الگوریتم؟

::::{admonition} راهنمایی‌ها و پاسخ
:class: dropdown

**راهنمایی ۱:** تابع یعنی رفتار بیرونی (ورودی ← خروجی) و الگوریتم یعنی روش محاسبه.

**راهنمایی ۲:** برای چند ورودی (مثلاً ۰، ۱، ۷، ۱۰) خروجی هر دو جعبه را بنویسید و بعد به عملی که درون هر جعبه انجام می‌شود نگاه کنید.

:::{admonition} پاسخ
:class: dropdown

**یک تابع** هستند، چون برای هر ورودی خروجی یکسان می‌دهند و از بیرون قابل تشخیص نیستند (<span dir="ltr">$x \mapsto 2x$</span>). اما **دو الگوریتم** هستند: یکی ضرب انجام می‌دهد و دیگری جمع. این همان تفاوت «تابع» و «الگوریتم» است.

**تمرین 2** · *Classify* · ★ (پایه)

برای هر مورد مشخص کنید تابع (کامل) است، تابع جزئی است یا اصلاً تابع نیست:

1. <span dir="ltr">$f : \mathbb{N} \to \mathbb{N}$</span> با <span dir="ltr">$f(x) = x - 1$</span>
2. <span dir="ltr">$g : \mathbb{Z} \to \mathbb{Z}$</span> با <span dir="ltr">$g(x) = x - 1$</span>
3. <span dir="ltr">$R = \{(x, y) \in \mathbb{N} \times \mathbb{N} \mid y^2 = x\}$</span> به‌عنوان رابطه‌ای از <span dir="ltr">$\mathbb{N}$</span> به <span dir="ltr">$\mathbb{N}$</span>
4. <span dir="ltr">$h(a, b) = a / b$</span> (تقسیم صحیح) روی <span dir="ltr">$\mathbb{N} \times \mathbb{N}$</span>

::::{admonition} راهنمایی‌ها و پاسخ
:class: dropdown

**راهنمایی ۱:** دو شرط تعریف را برای هر ورودی بسنجید: خروجی **وجود دارد**؟ و **یکتاست**؟

**راهنمایی ۲:** در <span dir="ltr">$\mathbb{N}$</span> عدد منفی نداریم؛ و در مورد ۳ بپرسید برای <span dir="ltr">$x = 2$</span> چند <span dir="ltr">$y$</span> طبیعی با <span dir="ltr">$y^2 = x$</span> هست؟

:::{admonition} پاسخ
:class: dropdown

1. **جزئی**: برای <span dir="ltr">$x = 0$</span> مقدار <span dir="ltr">$-1$</span> در <span dir="ltr">$\mathbb{N}$</span> نیست.
2. **کامل**.
3. **جزئی**: برای هر <span dir="ltr">$x$</span> حداکثر یک <span dir="ltr">$y \ge 0$</span> هست (یکتایی برقرار است)، اما برای <span dir="ltr">$x = 2$</span> هیچ <span dir="ltr">$y$</span> وجود ندارد.
4. **جزئی**: برای <span dir="ltr">$b = 0$</span> تعریف نشده است.

**تمرین 5** · *Small Implementation* · ★★ (متوسط)

برنامه‌ای بنویسید که با حلقه‌های تودرتو بررسی کند تابع <span dir="ltr">$f(x) = (3x + 1) \bmod 7$</span> روی <span dir="ltr">$\{0, \dots, 6\}$</span> دوسویی است یا نه.

::::{admonition} راهنمایی‌ها و پاسخ
:class: dropdown

**راهنمایی ۱:** یک‌به‌یک: برای هر جفت <span dir="ltr">$x_1 < x_2$</span> خروجی‌ها را مقایسه کنید. پوشا: برای هر <span dir="ltr">$y$</span> بگردید آیا <span dir="ltr">$x$</span>ی هست که <span dir="ltr">$f(x) = y$</span>.

**راهنمایی ۲:** دو متغیر پرچم (‎`injective`‎ و ‎`surjective`‎) با مقدار اولیهٔ ۱ بگیرید و هنگام دیدن یک تخلف آن‌ها را صفر کنید.

:::{admonition} پاسخ
:class: dropdown

```c
#include <stdio.h>

int main(void)
{
    /* f(x) = (3x + 1) mod 7 on {0, ..., 6} */
    int injective = 1, surjective = 1;

    for (int x1 = 0; x1 < 7; x1++)
        for (int x2 = x1 + 1; x2 < 7; x2++)
            if ((3 * x1 + 1) % 7 == (3 * x2 + 1) % 7) injective = 0;

    for (int y = 0; y < 7; y++) {
        int reached = 0;
        for (int x = 0; x < 7; x++)
            if ((3 * x + 1) % 7 == y) reached = 1;
        if (!reached) surjective = 0;
    }

    printf("injective : %s\n", injective ? "yes" : "no");
    printf("surjective: %s\n", surjective ? "yes" : "no");
    printf("bijective : %s\n", (injective && surjective) ? "yes" : "no");
    return 0;
}
```

خروجی:

```text
injective : yes
surjective: yes
bijective : yes
```

چون <span dir="ltr">$3$</span> و <span dir="ltr">$7$</span> نسبت به هم اول‌اند، ضرب در ۳ پیمانهٔ ۷ یک جایگشت است؛ جمع با ۱ هم آن را به هم نمی‌ریزد.

**تمرین 6** · *Prove / Explain* · ★★★ (چالشی)

نشان دهید اگر <span dir="ltr">$A$</span> متناهی با <span dir="ltr">$n$</span> عضو باشد، تعداد توابع <span dir="ltr">$f : A \to \{0, 1\}$</span> برابر <span dir="ltr">$2^n$</span> است. سپس توضیح دهید چرا با این مشاهده «ناشمارا بودن» توابع <span dir="ltr">$\mathbb{N} \to \{0, 1\}$</span> شهود معقولی دارد.

::::{admonition} راهنمایی‌ها و پاسخ
:class: dropdown

**راهنمایی ۱:** برای هر عضو <span dir="ltr">$A$</span> مستقل از بقیه دو انتخاب دارید.

**راهنمایی ۲:** یک تابع <span dir="ltr">$\mathbb{N} \to \{0, 1\}$</span> را مثل یک دنبالهٔ نامتناهی از صفر و یک ببینید؛ آیا می‌توان همهٔ این دنباله‌ها را در یک فهرست نامتناهی ردیف کرد؟ (ایدهٔ قطری)

:::{admonition} پاسخ
:class: dropdown

هر عضو <span dir="ltr">$A$</span> مستقلاً ۰ یا ۱ می‌گیرد، پس طبق اصل ضرب <span dir="ltr">$2 \cdot 2 \cdots 2 = 2^n$</span> تابع داریم. با بزرگ شدن <span dir="ltr">$n$</span> این عدد از خود <span dir="ltr">$n$</span> خیلی سریع‌تر رشد می‌کند. برای <span dir="ltr">$\mathbb{N}$</span> اگر فهرستی از همهٔ دنباله‌ها داشتیم، دنباله‌ای که <span dir="ltr">$i$</span>ام آن را برعکس <span dir="ltr">$i$</span>امین رقمِ <span dir="ltr">$i$</span>امین دنباله می‌گیریم در فهرست نیست (استدلال قطری کانتور). پس تعداد توابع **ناشمارا** است، در حالی که برنامه‌ها شمارا هستند.

**تمرین 8** · *Think* · ★★★ (چالشی)

فرض کنید تابع ‎`halts(p, x)`‎ وجود داشت. چگونه با آن می‌شد حدس کولاتس را حل کرد؟ (این نشان می‌دهد غیرمحاسبه‌پذیری فقط یک مسئلهٔ تئوری نیست.)

::::{admonition} راهنمایی‌ها و پاسخ
:class: dropdown

**راهنمایی ۱:** برنامه‌ای بسازید که برای <span dir="ltr">$n = 1, 2, 3, \dots$</span> با استفاده از ‎`halts`‎ بپرسد آیا برنامهٔ کولاتس روی <span dir="ltr">$n$</span> متوقف می‌شود.

**راهنمایی ۲:** چه وقت این برنامه متوقف می‌شود؟ سپس دوباره از ‎`halts`‎ دربارهٔ خود آن برنامه بپرسید.

:::{admonition} پاسخ
:class: dropdown

برنامهٔ <span dir="ltr">$R$</span> را می‌نویسیم: برای <span dir="ltr">$n = 1, 2, \dots$</span> اگر ‎`halts(collatz, n) == 0`‎ بود، توقف کن. برنامهٔ <span dir="ltr">$R$</span> **دقیقاً وقتی متوقف می‌شود** که یک مثال نقض برای حدس وجود داشته باشد. حالا ‎`halts(R)`‎ را حساب می‌کنیم: اگر ۱ باشد حدس **غلط** است و اگر ۰ باشد حدس **درست** است. پس وجود ‎`halts`‎ یک مسئلهٔ باز ریاضی را حل می‌کرد.

:class: dropdown

| فراخوانی که برگشت | ‎`n`‎ بعد از ‎`n = n + 1`‎ | ‎`result`‎ بعد از ضرب |

اگر ‎`n = n + 1`‎ بعد از ضرب بیاید، ضرب با مقدار خراب‌شدهٔ ‎`n`‎ (کم‌شده) انجام می‌شود و جواب اشتباه می‌شود.

برای هر متغیر مشخص کنید در کدام ناحیهٔ حافظه (Stack، Heap یا Static) است:

```c
int g = 10;

int f(int p) {
    int a = p + 1;
    static int s = 0;
    int *h = malloc(100);
    return a;
}
```

متغیرهای ‎`g`‎، ‎`p`‎، ‎`a`‎، ‎`s`‎، ‎`h`‎ و بلوکی که ‎`h`‎ به آن اشاره می‌کند را بررسی کنید.

:class: dropdown

**راهنمایی ۱:** متغیر سراسری و ‎`static`‎ طول عمرشان تا پایان برنامه است.

**راهنمایی ۲:** ‎`malloc`‎ از Heap می‌گیرد، ولی خود متغیر ‎`h`‎ یک متغیر محلی معمولی است.

:::{admonition} پاسخ
:class: dropdown

‎`g`‎: Static/Global. ‎`p`‎ و ‎`a`‎: Stack. ‎`s`‎: Static (با اینکه درون تابع تعریف شده). ‎`h`‎: Stack (خود اشاره‌گر). بلوکی که ‎`h`‎ به آن اشاره می‌کند: **Heap**.

**تمرین 12** · *Predict the Output* · ★★ (متوسط)

خروجی برنامهٔ زیر چیست و چرا؟

```c
#include <stdio.h>

void swap(int a, int b)
{
    int t = a;
    a = b;
    b = t;
}

int main(void)
{
    int x = 1, y = 2;
    swap(x, y);
    printf("%d %d\n", x, y);
    return 0;
}
```

::::{admonition} راهنمایی‌ها و پاسخ
:class: dropdown

**راهنمایی ۱:** آرگومان‌ها چگونه به پارامترها داده می‌شوند؟

**راهنمایی ۲:** ‎`a`‎ و ‎`b`‎ در کدام Stack Frame هستند و ‎`x`‎ و ‎`y`‎ در کدام؟

:::{admonition} پاسخ
:class: dropdown

**تمرین 18** · *Small Implementation* · ★★ (متوسط)

تابع بازگشتی مجموع ارقام یک عدد را یک بار بازگشتی و یک بار با حلقه بنویسید. آیا نسخهٔ بازگشتی شما دُم‌دار است؟ یک نسخهٔ دُم‌دار هم بنویسید.

::::{admonition} راهنمایی‌ها و پاسخ
:class: dropdown

**راهنمایی ۱:** <span dir="ltr">$\text{digits}(n) = (n \bmod 10) + \text{digits}(\lfloor n / 10 \rfloor)$</span> و برای <span dir="ltr">$n < 10$</span> خود <span dir="ltr">$n$</span> جواب است.

**راهنمایی ۲:** در بازگشت دُم‌دار فراخوانی باید آخرین کار باشد؛ حاصل جزئی را در یک پارامتر اضافه (‎`acc`‎) نگه دارید.

:::{admonition} پاسخ
:class: dropdown

```c
#include <stdio.h>

/* not tail recursive: the addition happens AFTER the call returns */
unsigned digits_rec(unsigned n)
{
    return n < 10 ? n : n % 10 + digits_rec(n / 10);
}

/* tail recursive: the partial sum travels in the parameter acc */
unsigned digits_tail(unsigned n, unsigned acc)
{
    return n == 0 ? acc : digits_tail(n / 10, acc + n % 10);
}

unsigned digits_loop(unsigned n)
{
    unsigned acc = 0;
    while (n != 0) {
        acc = acc + n % 10;
        n = n / 10;
    }
    return acc;
}

int main(void)
{
    printf("%u %u %u\n", digits_rec(9875), digits_tail(9875, 0), digits_loop(9875));
    return 0;
}
```

خروجی:

```text
29 29 29
```

نسخهٔ اول دُم‌دار **نیست**، چون بعد از برگشت هنوز جمع باقی است. نسخهٔ ‎`digits_tail`‎ دُم‌دار است و با حلقه هم‌ارز است (پارامتر ‎`acc`‎ همان متغیر حلقه می‌شود).

**تمرین 22** · *Prove* · ★★★ (چالشی)

با فرض اتحاد <span dir="ltr">$F(n) = F(k+1)\,F(n-k) + F(k)\,F(n-k-1)$</span> نشان دهید برای <span dir="ltr">$k \ge 1$</span>:

$$F(2k) = F(k)\big(F(k) + 2F(k-1)\big), \qquad F(2k+1) = \big(F(k) + F(k-1)\big)^2 + F(k)^2$$

و سپس <span dir="ltr">$F(2k+2)$</span> را بر حسب <span dir="ltr">$F(k-1)$</span> و <span dir="ltr">$F(k)$</span> بنویسید.

::::{admonition} راهنمایی‌ها و پاسخ
:class: dropdown

**راهنمایی ۱:** برای هر کدام <span dir="ltr">$n$</span> را مناسب انتخاب کنید: <span dir="ltr">$n = 2k$</span> و <span dir="ltr">$n = 2k+1$</span> و <span dir="ltr">$n = 2k+2$</span> (با همان <span dir="ltr">$k$</span>).

**راهنمایی ۲:** از <span dir="ltr">$F(k+1) = F(k) + F(k-1)$</span> و <span dir="ltr">$F(k+2) = F(k+1) + F(k) = 2F(k) + F(k-1)$</span> استفاده کنید.

:::{admonition} پاسخ
:class: dropdown

با <span dir="ltr">$n = 2k$</span> داریم <span dir="ltr">$n - k = k$</span>:
$$F(2k) = F(k+1)F(k) + F(k)F(k-1) = F(k)\big(F(k) + F(k-1) + F(k-1)\big) = F(k)\big(F(k) + 2F(k-1)\big).$$
با <span dir="ltr">$n = 2k+1$</span> داریم <span dir="ltr">$n - k = k + 1$</span>:
$$F(2k+1) = F(k+1)^2 + F(k)^2 = \big(F(k) + F(k-1)\big)^2 + F(k)^2.$$
با <span dir="ltr">$n = 2k+2$</span> داریم <span dir="ltr">$n - k = k + 2$</span>:
$$F(2k+2) = F(k+1)F(k+2) + F(k)F(k+1) = \big(F(k) + F(k-1)\big)\big(2F(k) + F(k-1)\big) + F(k)\big(F(k) + F(k-1)\big).$$

**تمرین 23** · *Trace* · ★ (پایه)

::::{admonition} راهنمایی‌ها و پاسخ
:class: dropdown

**راهنمایی ۱:** <span dir="ltr">$10 = 1010_2$</span>. در هر دور بیت پایین ‎`e`‎ را ببینید، بعد ‎`n`‎ را مربع کنید.

**راهنمایی ۲:** ‎`n`‎ به ترتیب <span dir="ltr">$2, 4, 16, 256$</span> است. فقط وقتی ‎`e`‎ فرد است ‎`res`‎ در ‎`n`‎ ضرب می‌شود.

:::{admonition} پاسخ
:class: dropdown

```c
#include <stdio.h>

typedef long long ll;
const ll MOD = 1000000007LL;

ll bin_pow(ll n, ll e)
{
    ll res = 1;
    while (e) {
        printf("res = %lld\n", res);
        if (e & 1) res = (res * n) % MOD;
        n = (n * n) % MOD;
        e >>= 1;
    }
    return res;
}

int main(void)
{
    printf("bin_pow(2, 10) = %lld\n", bin_pow(2, 10));
    return 0;
}
```

**تمرین 26** · *Implementation* · ★★ (متوسط)

تابع ‎`bin_pow`‎ درسنامه پیمانهٔ ثابت <span dir="ltr">$10^9 + 7$</span> دارد. آن را به ‎`mod_power(a, n, m)`‎ تعمیم دهید (با <span dir="ltr">$m < 2^{32}$</span> تا ضرب‌ها سرریز نکنند) و یک بار با حلقه و یک بار بازگشتی بنویسید. سپس <span dir="ltr">$7^{10^{18}} \bmod 1000003$</span> را با هر دو حساب کنید.

::::{admonition} راهنمایی‌ها و پاسخ
:class: dropdown

**راهنمایی ۱:** بعد از **هر** ضرب باقی‌مانده بگیرید تا عدد بزرگ نشود. مقدار اولیهٔ جواب را ‎`1 % m`‎ بگیرید تا <span dir="ltr">$m = 1$</span> هم درست کار کند.

**راهنمایی ۲:** نسخهٔ حلقه همان ‎`bin_pow`‎ است؛ نسخهٔ بازگشتی همان ‎`bin_pow_rec`‎ با ‎`half`‎ و مربع آن.

:::{admonition} پاسخ
:class: dropdown

```c
#include <stdio.h>

typedef unsigned long long ull;

/* a^n mod m, assuming m < 2^32 so that every product fits in 64 bits */
ull mod_power(ull a, ull n, ull m)            /* iterative */
{
    ull res = 1 % m;
    a %= m;
    while (n) {
        if (n & 1) res = res * a % m;
        a = a * a % m;
        n >>= 1;
    }
    return res;
}

ull mod_power_rec(ull a, ull n, ull m)        /* recursive */
{
    if (n == 0) return 1 % m;
    ull half = mod_power_rec(a, n / 2, m);
    ull r = half * half % m;
    return n % 2 == 1 ? r * (a % m) % m : r;
}

int main(void)
{
    printf("2^10 mod 1000         = %llu\n", mod_power(2, 10, 1000));
    printf("3^200 mod 1000000007  = %llu / %llu\n",
           mod_power(3, 200, 1000000007ULL), mod_power_rec(3, 200, 1000000007ULL));
    printf("7^(10^18) mod 1000003 = %llu / %llu\n",
           mod_power(7, 1000000000000000000ULL, 1000003ULL),
           mod_power_rec(7, 1000000000000000000ULL, 1000003ULL));
    return 0;
}
```

خروجی:

```text
2^10 mod 1000         = 24
3^200 mod 1000000007  = 136318165 / 136318165
7^(10^18) mod 1000003 = 32502 / 32502
```

<span dir="ltr">$7^{10^{18}} \bmod 1000003 = 32502$</span> و هر دو نسخه جواب یکسان می‌دهند. حلقه حدود ۶۰ دور و بازگشت حدود ۶۰ فراخوانی دارد.

**تمرین 29** · *Prove* · ★★★ (چالشی)

اتحاد <span dir="ltr">$F(n) = F(k+1)\,F(n-k) + F(k)\,F(n-k-1)$</span> را برای <span dir="ltr">$n$</span> ثابت و <span dir="ltr">$0 \le k \le n-1$</span> با استقرا روی <span dir="ltr">$k$</span> ثابت کنید.

::::{admonition} راهنمایی‌ها و پاسخ
:class: dropdown

**راهنمایی ۱:** پایه <span dir="ltr">$k = 0$</span> است: سمت راست چه می‌شود؟

**راهنمایی ۲:** در گام استقرا <span dir="ltr">$F(n-k) = F(n-k-1) + F(n-k-2)$</span> را جای‌گذاری کنید و جمله‌های دارای <span dir="ltr">$F(n-k-1)$</span> را کنار هم بیاورید.

:::{admonition} پاسخ
:class: dropdown

**پایه** <span dir="ltr">$k = 0$</span>: <span dir="ltr">$F(1)F(n) + F(0)F(n-1) = F(n)$</span> ✓. **گام**: با فرض درستی برای <span dir="ltr">$k$</span> (و <span dir="ltr">$k + 2 \le n$</span>):
$$F(n) = F(k+1)\big(F(n-k-1) + F(n-k-2)\big) + F(k)F(n-k-1) = \big(F(k+1) + F(k)\big)F(n-k-1) + F(k+1)F(n-k-2),$$
که برابر <span dir="ltr">$F(k+2)F(n-k-1) + F(k+1)F(n-k-2)$</span> یعنی حکم برای <span dir="ltr">$k+1$</span> است. ∎

**تمرین 43** · *Traverse* · ★★ (متوسط)

با کمک ‎`adt.h`‎ یک BFS بنویسید که درخت بازگشت ‎`fib(5)`‎ را سطح‌به‌سطح بپیماید و تعداد گره‌های هر سطح را چاپ کند. عرض بیشینه چند است؟

::::{admonition} راهنمایی‌ها و پاسخ
:class: dropdown

**راهنمایی ۱:** هر گره را به‌صورت جفت (آرگومان، شمارهٔ سطح) در Queue بگذارید.

**راهنمایی ۲:** وقتی شمارهٔ سطح عوض شد، تعداد سطح قبلی را چاپ کنید و شمارنده را صفر کنید.

:::{admonition} پاسخ
:class: dropdown

```c
#include <stdio.h>
#include "adt.h"

int main(void)
{
    int level = 0, count = 0, widest = 0;

    q_push(5); q_push(0);                  /* node fib(5) at level 0 */
    while (!q_empty()) {
        int n = q_pop();
        int lv = q_pop();

        if (lv != level) {                 /* the previous level is finished */
            printf("level %d: %d nodes\n", level, count);
            if (count > widest) widest = count;
            level = lv;
            count = 0;
        }
        count++;

        if (n >= 2) {                      /* two children: fib(n-1) and fib(n-2) */
            q_push(n - 1); q_push(lv + 1);
            q_push(n - 2); q_push(lv + 1);
        }
    }
    printf("level %d: %d nodes\n", level, count);
    if (count > widest) widest = count;
    printf("maximum width = %d\n", widest);
    return 0;
}
```

```text
level 0: 1 nodes
level 1: 2 nodes
level 2: 4 nodes
level 3: 6 nodes
level 4: 2 nodes
maximum width = 6
```

## ک) ماکروها

**تمرین 46** · *Fix the Macro* · ★★ (متوسط)

ماکروهای ‎`MAX_BAD(a,b)`‎ به شکل ‎`a > b ? a : b`‎ و ‎`TWICE(x)`‎ به شکل ‎`2 * x`‎ و ‎`CUBE_BAD(x)`‎ به شکل ‎`x * x * x`‎ را روی ‎`MAX_BAD(2,1)+10`‎ و ‎`TWICE(1+2)`‎ و ‎`CUBE_BAD(1+2)`‎ بسط دهید و بگویید چه مقدارهایی به‌دست می‌آید. نسخهٔ امن هر کدام را بنویسید. سپس نشان دهید با ‎`CUBE(i++)`‎ حتی نسخهٔ امن هم مشکل دارد.

::::{admonition} راهنمایی‌ها و پاسخ
:class: dropdown

**راهنمایی ۱:** پیش‌پردازنده فقط **متن** را جایگزین می‌کند؛ بسط‌ها را بنویسید و بعد اولویت عملگرها را اعمال کنید.

**راهنمایی ۲:** قاعدهٔ پرانتزگذاری: هر پارامتر و کل بدنه. برای آخرین قسمت ببینید ‎`i++`‎ چند بار در متن می‌آید.

:::{admonition} پاسخ
:class: dropdown

```c
#include <stdio.h>

#define MAX_BAD(a, b)  a > b ? a : b
#define MAX(a, b)      ((a) > (b) ? (a) : (b))
#define TWICE(x)       2 * x
#define CUBE_BAD(x)    x * x * x
#define CUBE(x)        ((x) * (x) * (x))

int main(void)
{
    printf("MAX_BAD(2, 1) + 10 = %d   (expected 12)\n", MAX_BAD(2, 1) + 10);
    printf("MAX(2, 1) + 10     = %d\n", MAX(2, 1) + 10);
    printf("TWICE(1 + 2)       = %d   (expected 6)\n", TWICE(1 + 2));
    printf("CUBE_BAD(1 + 2)    = %d   (expected 27)\n", CUBE_BAD(1 + 2));
    printf("CUBE(1 + 2)        = %d\n", CUBE(1 + 2));
    return 0;
}
```

خروجی:

```text
MAX_BAD(2, 1) + 10 = 2   (expected 12)
MAX(2, 1) + 10     = 12
TWICE(1 + 2)       = 4   (expected 6)
CUBE_BAD(1 + 2)    = 7   (expected 27)
CUBE(1 + 2)        = 27
```

بسط‌ها: ‎`MAX_BAD(2,1)+10`‎ ← ‎`2 > 1 ? 2 : 1 + 10`‎ (جمع در شاخهٔ غلط می‌افتد؛ نتیجه ۲ نه ۱۲)، ‎`TWICE(1+2)`‎ ← ‎`2 * 1 + 2 = 4`‎، ‎`CUBE_BAD(1+2)`‎ ← ‎`1 + 2 * 1 + 2 * 1 + 2 = 7`‎. نسخهٔ امن: ‎`((a) > (b) ? (a) : (b))`‎ و ‎`(2 * (x))`‎ و ‎`((x) * (x) * (x))`‎. اما ‎`CUBE(i++)`‎ به ‎`((i++) * (i++) * (i++))`‎ بسط می‌یابد: ‎`i`‎ سه بار در یک عبارت تغییر می‌کند و رفتار برنامه **تعریف‌نشده** است. ماکرو هرگز نباید آرگومان با اثر جانبی بگیرد.

## ل) مسئلهٔ جامع و چالش‌های پژوهشی

**تمرین 47** · *Final Challenge: Tower of Hanoi* · ★★★ (چالشی)

**برج‌های هانوی** (ادوار لوکا، ۱۸۸۳): سه میله و <span dir="ltr">$n$</span> دیسک با اندازه‌های متفاوت روی میلهٔ اول (بزرگ‌تر پایین‌تر) داریم. می‌خواهیم همهٔ دیسک‌ها را به میلهٔ سوم ببریم، با دو قید: در هر حرکت فقط **یک دیسک** جابه‌جا شود و **هیچ دیسک بزرگی روی کوچک‌تر** قرار نگیرد.

۱. تابع ‎`hanoi(n, from, to, via)`‎ را بنویسید که هر حرکت را چاپ کند. حالت پایه چیست؟ اندازهٔ کاهنده کدام است؟
۲. نشان دهید تعداد حرکات <span dir="ltr">$T(n) = 2T(n-1) + 1$</span> و <span dir="ltr">$T(n) = 2^n - 1$</span> است. درخت بازگشت چه شکلی دارد؟ ارتفاع و تعداد گره‌های آن چقدر است؟
۳. با شمارندهٔ سراسری، تعداد فراخوانی‌ها و بیشینهٔ عمق را برای چند <span dir="ltr">$n$</span> چاپ کنید و با تحلیل خود بسنجید.
۴. ‎`hanoi`‎ را با یک حلقه و Stack صریح بنویسید که **همان ترتیب حرکات** را چاپ کند. چه چیزی را در Stack می‌گذارید؟
۵. آیا می‌توان با Queue (BFS) همان ترتیب حرکات را گرفت؟ دلیل بیاورید.

::::{admonition} راهنمایی‌ها و پاسخ
:class: dropdown

**راهنمایی ۱:** حالت پایه <span dir="ltr">$n = 0$</span> است (هیچ کاری نکن). بازگشت دو بار با <span dir="ltr">$n-1$</span> فراخوانی می‌شود و یک حرکت چاپی **بین** آن‌هاست (درون‌ترتیب درخت بازگشت). برای حذف بازگشت، به‌جای «انتقال <span dir="ltr">$n$</span> دیسک» سه کار «انتقال <span dir="ltr">$n-1$</span>»، «حرکت دیسک <span dir="ltr">$n$</span>» و «انتقال <span dir="ltr">$n-1$</span>» را به ترتیب **معکوس** push کنید و «حرکت تکی» را با یک نشانه از «انتقال» جدا کنید.

**راهنمایی ۲:** BFS گره‌ها را سطح‌به‌سطح می‌بیند، پس حرکت دیسک بزرگ (ریشه) را پیش از حرکت‌های دیسک‌های کوچک‌تر (که باید قبلش انجام شوند) می‌بیند. ترتیب درست حرکت‌ها «درون‌ترتیب» درخت است که نه DFS پیش‌ترتیب است و نه سطحی.

:::{admonition} پاسخ
:class: dropdown

**۱) نسخهٔ بازگشتی** (با شمارنده‌ها برای قسمت ۳):

```c
#include <stdio.h>

static unsigned long moves, calls, depth, max_depth;

void hanoi(int n, char from, char to, char via, int verbose)
{
    calls++;
    depth++;
    if (depth > max_depth) max_depth = depth;

    if (n > 0) {
        hanoi(n - 1, from, via, to, verbose);        /* move n-1 disks out of the way */
        moves++;
        if (verbose) printf("move disk %d: %c -> %c\n", n, from, to);
        hanoi(n - 1, via, to, from, verbose);        /* put them on top of disk n     */
    }
    depth--;
}

int main(void)
{
    hanoi(3, 'A', 'C', 'B', 1);

    printf("\n  n     moves     calls   max depth\n");
    int tests[] = {1, 2, 3, 10, 20};
    for (int i = 0; i < 5; i++) {
        moves = calls = depth = max_depth = 0;
        hanoi(tests[i], 'A', 'C', 'B', 0);
        printf("%3d  %8lu  %8lu  %10lu\n", tests[i], moves, calls, max_depth);
    }
    return 0;
}
```

خروجی:

```text
move disk 1: A -> C
move disk 2: A -> B
move disk 1: C -> B
move disk 3: A -> C
move disk 1: B -> A
move disk 2: B -> C
move disk 1: A -> C

  n     moves     calls   max depth
  1         1         3           2
  2         3         7           3
  3         7        15           4
 10      1023      2047          11
 20   1048575   2097151          21
```

حالت پایه <span dir="ltr">$n = 0$</span> و اندازهٔ کاهنده <span dir="ltr">$n$</span> است.

**۲) تحلیل:** برای انتقال <span dir="ltr">$n$</span> دیسک: ابتدا <span dir="ltr">$n-1$</span> دیسک (T(n−1) حرکت)، یک حرکت برای بزرگ‌ترین دیسک و دوباره <span dir="ltr">$n-1$</span> دیسک؛ پس <span dir="ltr">$T(n) = 2T(n-1) + 1$</span> با <span dir="ltr">$T(0) = 0$</span>. با بسط: <span dir="ltr">$T(n) = 2^n - 1$</span> (استقرا: <span dir="ltr">$2(2^{n-1} - 1) + 1 = 2^n - 1$</span>). درخت بازگشت یک **درخت دودویی کامل** است؛ با پایهٔ <span dir="ltr">$n = 0$</span> ارتفاع آن <span dir="ltr">$n$</span> و تعداد گره‌های آن <span dir="ltr">$2^{n+1} - 1$</span> است (جدول بالا: برای <span dir="ltr">$n = 10$</span> تعداد ۲۰۴۷). بیشینهٔ عمق Stack برابر <span dir="ltr">$n + 1$</span> است؛ یعنی زمان نمایی و حافظه خطی، و <span dir="ltr">$n = 64$</span> دیسک حدود <span dir="ltr">$1.8 \times 10^{19}$</span> حرکت می‌خواهد.

**۴) حذف بازگشت:** هر «کار» را به‌صورت (n, from, to, via, is_move) در Stack می‌گذاریم و کارها را به ترتیب معکوس push می‌کنیم:

```c
#include <stdio.h>
#include "adt.h"

/* a task is (n, from, to, via, is_move) stored as five ints */
void push_task(int n, char from, char to, char via, int is_move)
{
    s_push(n); s_push(from); s_push(to); s_push(via); s_push(is_move);
}

void hanoi_iter(int n, char from, char to, char via)
{
    push_task(n, from, to, via, 0);
    while (!s_empty()) {
        int is_move = s_pop();
        char v = (char)s_pop(), t = (char)s_pop(), f = (char)s_pop();
        int k = s_pop();

        if (is_move) {
            printf("move disk %d: %c -> %c\n", k, f, t);
        } else if (k > 0) {
            push_task(k - 1, v, t, f, 0);      /* done last : n-1 disks from via to to  */
            push_task(k, f, t, v, 1);          /* done second: a single move            */
            push_task(k - 1, f, v, t, 0);      /* done first: n-1 disks from from to via */
        }
    }
}

int main(void)
{
    hanoi_iter(3, 'A', 'C', 'B');
    return 0;
}
```

**۵) چرا BFS نه؟** ترتیب حرکات «درون‌ترتیب» درخت است: ابتدا زیردرخت چپ، بعد گرهٔ جاری و بعد زیردرخت راست. BFS ریشه (حرکت بزرگ‌ترین دیسک) را قبل از حرکت‌های زیردرخت چپ می‌بیند، پس ترتیب را به‌هم می‌ریزد. شرط دوم BFS هم نقض می‌شود: اعتبار هر حرکت به موقعیت دیسک‌ها پس از **حرکت‌های قبلی** بستگی دارد.

تابع <span dir="ltr">$\text{BB}(n)$</span> (Busy Beaver) را جست‌وجو کنید. چرا از آکرمان هم سریع‌تر رشد می‌کند و چرا محاسبه‌پذیر نیست؟ این با مسئلهٔ توقف چه ارتباطی دارد؟

::::{admonition} راهنمایی‌ها و پاسخ
:class: dropdown

**راهنمایی ۱:** <span dir="ltr">$\text{BB}(n)$</span> بیشینهٔ تعداد گام‌هایی است که یک ماشین تورینگ <span dir="ltr">$n$</span> حالته که **متوقف می‌شود** می‌تواند بردارد.

**راهنمایی ۲:** اگر <span dir="ltr">$\text{BB}$</span> محاسبه‌پذیر بود، با اجرای هر ماشین به‌اندازهٔ <span dir="ltr">$\text{BB}(n)$</span> گام چه می‌شد فهمید؟

:::{admonition} پاسخ
:class: dropdown

اگر <span dir="ltr">$\text{BB}$</span> محاسبه‌پذیر بود، برای هر ماشین <span dir="ltr">$n$</span> حالته آن را <span dir="ltr">$\text{BB}(n)$</span> گام اجرا می‌کردیم؛ اگر تا آن زمان تمام نشده باشد، هرگز تمام نمی‌شود. این **مسئلهٔ توقف را حل می‌کرد**، که ناممکن است. پس <span dir="ltr">$\text{BB}$</span> محاسبه‌پذیر نیست و از **هر** تابع محاسبه‌پذیر (از جمله آکرمان) سریع‌تر رشد می‌کند، چون در غیر این صورت با همان تابع محاسبه‌پذیر به‌عنوان کران بالا می‌شد توقف را تشخیص داد.
