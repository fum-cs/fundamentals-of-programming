````python
# ============================================================
# جعبه‌ابزار تصویرسازی (Visualization Toolkit) - فقط اجرا کنید و رد شوید
# پخش‌کننده‌ی تعاملی الگوریتم‌ها: کد، متغیرها، توضیح هر گام و خروجی
# ============================================================
import html as _html, json, re, uuid, hashlib
from IPython.display import HTML, display

_CF_CSS = r''':host{display:block;margin:1.1rem 0}
*{box-sizing:border-box}
.card{--bg:#ffffff;--panel:#f5f6fb;--stage:#f9fafe;--dot:#dfe3f0;--bd:#e2e5f0;--tx:#1a1f36;--mut:#68718d;--acc:#4f46e5;--acc2:#7c3aed;--acc-soft:#ecebff;--hl:rgba(79,70,229,.11);--shadow:0 10px 30px -18px rgba(30,30,80,.35);
--code-bg:#fcfcff;--ln:#a3abc2;--k-kw:#a626a4;--k-ty:#0b7285;--k-num:#c2410c;--k-str:#15803d;--k-cm:#8a93ab;--k-fn:#2563eb;--k-pp:#9a3412;--k-tx:#1f2540;
--i-bg:#eef0f8;--i-bd:#cfd5e8;--i-tx:#2a3050;
--cur-bg:#fde68a;--cur-bd:#d97706;--cur-tx:#5b3a00;--cur-gl:rgba(217,119,6,.22);
--cmp-bg:#bae6fd;--cmp-bd:#0284c7;--cmp-tx:#073b5a;
--ok-bg:#bbf7d0;--ok-bd:#16a34a;--ok-tx:#0b3d1d;
--min-bg:#fecaca;--min-bd:#dc2626;--min-tx:#5e0d0d;
--key-bg:#e9d5ff;--key-bd:#9333ea;--key-tx:#3b0a63;
--sorted-bg:#c7d2fe;--sorted-bd:#4f46e5;--sorted-tx:#1e1b6b;
--vis-bg:#dfe5f5;--vis-bd:#9aa6c9;--vis-tx:#3a4468;
--out-bg:#0f1424;--out-fg:#86efac;--out-bd:#1d2540}
.card[data-theme=dark]{--bg:#12162a;--panel:#1a1f38;--stage:#0f1325;--dot:#222a49;--bd:#2a3150;--tx:#e8ebf7;--mut:#9aa4c4;--acc:#8b93ff;--acc2:#c4a1ff;--acc-soft:#262c5a;--hl:rgba(139,147,255,.16);--shadow:0 10px 30px -18px rgba(0,0,0,.8);
--code-bg:#0e1224;--ln:#566087;--k-kw:#e19bff;--k-ty:#5fd3e8;--k-num:#ffb27a;--k-str:#7ee2a0;--k-cm:#6c7699;--k-fn:#8db4ff;--k-pp:#ffb08a;--k-tx:#e3e7fa;
--i-bg:#1d2340;--i-bd:#39426b;--i-tx:#e6e9fb;
--cur-bg:#6b4b06;--cur-bd:#fbbf24;--cur-tx:#ffeaa8;--cur-gl:rgba(251,191,36,.25);
--cmp-bg:#0b3a58;--cmp-bd:#38bdf8;--cmp-tx:#d4efff;
--ok-bg:#0f4a2a;--ok-bd:#4ade80;--ok-tx:#d3ffe3;
--min-bg:#5a1717;--min-bd:#f87171;--min-tx:#ffd9d9;
--key-bg:#3f1a63;--key-bd:#c084fc;--key-tx:#f0e0ff;
--sorted-bg:#262c7a;--sorted-bd:#818cf8;--sorted-tx:#e0e3ff;
--vis-bg:#252c4a;--vis-bd:#4b5680;--vis-tx:#b4bddc;
--out-bg:#070a16;--out-fg:#86efac;--out-bd:#1d2540}
.card{font-family:'Vazirmatn',system-ui,-apple-system,'Segoe UI',Roboto,sans-serif;color:var(--tx);background:var(--bg);border:1px solid var(--bd);border-radius:18px;overflow:hidden;box-shadow:var(--shadow);outline:none;direction:rtl}
.card:focus-visible{box-shadow:0 0 0 3px var(--acc-soft),0 0 0 1px var(--acc)}
.card.full{position:fixed;inset:0;z-index:99999;border-radius:0;overflow:auto;display:flex;flex-direction:column}
.card.full .stage{flex:1}
.hd{display:flex;align-items:center;justify-content:space-between;gap:10px;padding:10px 14px;border-bottom:1px solid var(--bd);background:linear-gradient(180deg,var(--panel),var(--bg))}
.ttl{display:flex;align-items:center;gap:9px;min-width:0}
.dot{width:10px;height:10px;border-radius:50%;background:linear-gradient(135deg,var(--acc),var(--acc2));box-shadow:0 0 0 4px var(--acc-soft);flex:none}
.tt{font:700 14px/1.4 'Vazirmatn',system-ui,sans-serif;overflow:hidden;text-overflow:ellipsis;white-space:nowrap}
.hr{display:flex;align-items:center;gap:8px;flex:none}
.bdg{font:700 10.5px/1 'Vazirmatn',system-ui,sans-serif;letter-spacing:.04em;color:var(--acc);background:var(--acc-soft);padding:5px 10px;border-radius:99px}
.ib{display:grid;place-items:center;width:30px;height:30px;border-radius:9px;border:1px solid var(--bd);background:var(--bg);color:var(--mut);cursor:pointer;padding:0;transition:.15s}
.ib:hover{color:var(--acc);border-color:var(--acc);background:var(--acc-soft)}
.ib svg{width:15px;height:15px;fill:currentColor}
.stage{position:relative;padding:18px 12px 10px;background-color:var(--stage);background-image:radial-gradient(var(--dot) 1px,transparent 1.2px);background-size:18px 18px;border-bottom:1px solid var(--bd);overflow-x:auto}
.row{display:flex;justify-content:center;gap:var(--g,10px);min-width:max-content;margin:0 auto;direction:ltr}
.cell{display:flex;flex-direction:column;align-items:center;width:var(--bw,56px)}
.box{width:var(--bw,56px);height:var(--s,56px);display:grid;place-items:center;border:2px solid var(--i-bd);border-radius:calc(var(--s,56px)*.22);background:var(--i-bg);color:var(--i-tx);font:700 var(--fs,19px)/1 'JetBrains Mono',ui-monospace,Consolas,monospace;direction:ltr;transition:background-color .25s,border-color .25s,color .25s,box-shadow .25s,transform .25s,opacity .25s;white-space:nowrap}
.box[data-c=cur]{background:var(--cur-bg);border-color:var(--cur-bd);color:var(--cur-tx);box-shadow:0 0 0 5px var(--cur-gl);transform:translateY(-3px)}
.box[data-c=cmp]{background:var(--cmp-bg);border-color:var(--cmp-bd);color:var(--cmp-tx)}
.box[data-c=ok]{background:var(--ok-bg);border-color:var(--ok-bd);color:var(--ok-tx)}
.box[data-c=min]{background:var(--min-bg);border-color:var(--min-bd);color:var(--min-tx)}
.box[data-c=key]{background:var(--key-bg);border-color:var(--key-bd);color:var(--key-tx)}
.box[data-c=sorted]{background:var(--sorted-bg);border-color:var(--sorted-bd);color:var(--sorted-tx)}
.box[data-c=vis]{background:var(--vis-bg);border-color:var(--vis-bd);color:var(--vis-tx)}
.box[data-c=dim]{background:transparent;border-style:dashed;opacity:.45}
.idx{margin-top:7px;font:600 11px/1 'JetBrains Mono',ui-monospace,Consolas,monospace;color:var(--mut);direction:ltr;white-space:nowrap}
.tags{display:flex;flex-direction:column;align-items:center;gap:4px;margin-top:5px;min-height:46px}
.ptr{position:relative;font:700 11.5px/1 'JetBrains Mono',ui-monospace,Consolas,monospace;color:#fff;background:var(--acc);padding:4px 9px 4px;border-radius:7px;animation:pin .35s cubic-bezier(.3,1.5,.5,1);direction:ltr}
.ptr:first-child::before{content:"";position:absolute;top:-5px;left:50%;margin-left:-5px;border:5px solid transparent;border-top:0;border-bottom:5px solid var(--acc)}
@keyframes pin{from{opacity:0;transform:translateY(8px) scale(.7)}to{opacity:1;transform:none}}
.grid{display:grid;justify-content:center;align-items:center;gap:8px;direction:ltr;margin:0 auto;width:max-content}
.hdr{font:600 11.5px/1 'JetBrains Mono',ui-monospace,Consolas,monospace;color:var(--mut);text-align:center;padding:4px 6px;border-radius:7px;transition:.2s;white-space:nowrap}
.hdr.on{color:#fff;background:var(--acc)}
.lg{display:flex;flex-wrap:wrap;gap:6px 16px;padding:9px 14px;border-bottom:1px solid var(--bd);background:var(--bg);font:500 12px/1 'Vazirmatn',system-ui,sans-serif;color:var(--mut)}
.lg span{display:inline-flex;align-items:center;gap:6px}
.lg i{width:13px;height:13px;border-radius:4px;border:2px solid var(--i-bd);background:var(--i-bg)}
.lg i[data-c=cur]{background:var(--cur-bg);border-color:var(--cur-bd)}.lg i[data-c=cmp]{background:var(--cmp-bg);border-color:var(--cmp-bd)}
.lg i[data-c=ok]{background:var(--ok-bg);border-color:var(--ok-bd)}.lg i[data-c=min]{background:var(--min-bg);border-color:var(--min-bd)}
.lg i[data-c=key]{background:var(--key-bg);border-color:var(--key-bd)}.lg i[data-c=sorted]{background:var(--sorted-bg);border-color:var(--sorted-bd)}
.lg i[data-c=vis]{background:var(--vis-bg);border-color:var(--vis-bd)}.lg i[data-c=dim]{background:transparent;border-style:dashed;opacity:.6}
.split{display:grid;grid-template-columns:minmax(0,1.12fr) minmax(0,1fr)}
.card.narrow .split{grid-template-columns:minmax(0,1fr)}
.codep{border-left:1px solid var(--bd);min-width:0;direction:ltr;background:var(--code-bg)}
.card.narrow .codep{border-left:0;border-bottom:1px solid var(--bd)}
.chd{display:flex;align-items:center;gap:9px;padding:9px 14px;border-bottom:1px solid var(--bd);font:700 10.5px/1 'JetBrains Mono',ui-monospace,Consolas,monospace;letter-spacing:.08em;color:var(--mut)}
.chd s{width:9px;height:9px;border-radius:50%;background:#ff5f57;text-decoration:none;box-shadow:15px 0 0 #febc2e,30px 0 0 #28c840;margin-right:34px;flex:none}
.cbody{position:relative;padding:8px 0 10px;overflow:auto;max-height:300px;overscroll-behavior:contain;font:500 13px/22px 'JetBrains Mono',ui-monospace,Consolas,monospace;color:var(--k-tx)}
.lines{position:relative;min-width:max-content}
.hlb{position:absolute;left:0;right:0;top:0;height:22px;background:var(--hl);border-left:3px solid var(--acc);transition:transform .35s cubic-bezier(.4,0,.2,1),height .35s cubic-bezier(.4,0,.2,1);z-index:0}
.ln{position:relative;display:flex;height:22px;white-space:pre;z-index:1}
.no{flex:none;width:40px;padding-right:12px;text-align:right;color:var(--ln);user-select:none}
.ln.on .no{color:var(--acc);font-weight:700}
.tx{flex:1;padding-right:16px}
.k{color:var(--k-kw)}.t{color:var(--k-ty)}.n{color:var(--k-num)}.s{color:var(--k-str)}.c{color:var(--k-cm);font-style:italic}.f{color:var(--k-fn)}.p{color:var(--k-pp)}
.info{display:flex;flex-direction:column;gap:13px;padding:13px 15px;min-width:0;background:var(--bg)}
.bt{margin-bottom:7px;font:700 11.5px/1 'Vazirmatn',system-ui,sans-serif;color:var(--mut)}
.vars{display:flex;flex-wrap:wrap;gap:7px;min-height:32px}
.chip{direction:ltr;display:flex;align-items:baseline;gap:7px;padding:5px 11px;border-radius:10px;background:var(--panel);border:1px solid var(--bd);font:600 13px/1.25 'JetBrains Mono',ui-monospace,Consolas,monospace}
.chip b{color:var(--mut);font-weight:600}.chip span{color:var(--tx);font-weight:700;min-width:1ch}
.chip.na{opacity:.45}.chip.chg{animation:pop .7s ease-out}
@keyframes pop{0%{background:var(--acc-soft);border-color:var(--acc);transform:scale(1.08)}100%{transform:scale(1)}}
.note{min-height:70px;padding:11px 13px;border-radius:12px;background:var(--panel);border:1px solid var(--bd);font:500 13.5px/1.75 'Vazirmatn',system-ui,sans-serif}
.note.in{animation:rise .35s ease-out}
@keyframes rise{from{opacity:0;transform:translateY(5px)}to{opacity:1;transform:none}}
.note code,.cap code{font:600 12.5px 'JetBrains Mono',ui-monospace,Consolas,monospace;color:var(--acc);background:var(--acc-soft);padding:1px 6px;border-radius:6px;direction:ltr;unicode-bidi:isolate}
.out pre{margin:0;direction:ltr;text-align:left;min-height:56px;max-height:150px;overflow:auto;white-space:pre-wrap;padding:10px 12px;border-radius:12px;background:var(--out-bg);color:var(--out-fg);border:1px solid var(--out-bd);font:500 12.5px/1.6 'JetBrains Mono',ui-monospace,Consolas,monospace}
.out pre::after{content:"\2588";margin-left:2px;animation:blink 1s steps(1) infinite}
@keyframes blink{50%{opacity:0}}
/* ---------- redesigned control dock ---------- */
.dock{display:flex;align-items:center;gap:14px;flex-wrap:wrap;padding:12px 14px;background:var(--panel);direction:ltr}
.tp{display:flex;align-items:center;gap:4px;padding:4px;border-radius:99px;background:var(--bg);border:1px solid var(--bd)}
.tb{display:grid;place-items:center;width:32px;height:32px;padding:0;border:0;border-radius:50%;background:transparent;color:var(--tx);cursor:pointer;transition:.15s}
.tb svg{width:16px;height:16px;fill:currentColor}
.tb:hover{background:var(--acc-soft);color:var(--acc)}
.tb:disabled{opacity:.3;cursor:default;background:transparent;color:var(--tx)}
.tb:focus-visible{outline:2px solid var(--acc);outline-offset:1px}
.tb.go{width:42px;height:42px;background:linear-gradient(135deg,var(--acc),var(--acc2));color:#fff;box-shadow:0 8px 16px -8px var(--acc)}
.tb.go svg{width:19px;height:19px}.tb.go:hover{filter:brightness(1.1);color:#fff}
.tl{flex:1;min-width:150px;display:flex;flex-direction:column;gap:5px}
.tinfo{display:flex;justify-content:space-between;font:600 11.5px/1 'Vazirmatn',system-ui,sans-serif;color:var(--mut);direction:rtl}
.tinfo b{font-family:'JetBrains Mono',ui-monospace,monospace;color:var(--tx);direction:ltr;unicode-bidi:isolate}
.track{position:relative;height:22px;cursor:pointer;touch-action:none;outline:none}
.rail{position:absolute;left:0;right:0;top:8px;height:6px;border-radius:99px;background:var(--bd);overflow:hidden}
.fill{height:100%;width:0;border-radius:inherit;background:linear-gradient(90deg,var(--acc),var(--acc2));transition:width .2s}
.tick{position:absolute;top:5px;width:2px;height:12px;margin-left:-1px;border-radius:2px;background:var(--bg);opacity:.9;pointer-events:none}
.thumb{position:absolute;top:3px;width:16px;height:16px;margin-left:-8px;border-radius:50%;background:#fff;border:3px solid var(--acc);box-shadow:0 2px 8px rgba(0,0,0,.28);transition:left .2s}
.track:focus-visible .thumb{box-shadow:0 0 0 4px var(--acc-soft)}
.sp{display:flex;padding:3px;border-radius:99px;background:var(--bg);border:1px solid var(--bd)}
.sp button{border:0;background:transparent;color:var(--mut);padding:6px 9px;border-radius:99px;font:700 11.5px/1 'JetBrains Mono',ui-monospace,monospace;cursor:pointer;transition:.15s}
.sp button:hover{color:var(--acc)}.sp button.on{background:var(--acc);color:#fff}
.ib.tg.on{background:var(--acc-soft);border-color:var(--acc);color:var(--acc)}
.card.narrow .dock{gap:10px}.card.narrow .tl{order:5;flex-basis:100%}
@media (prefers-reduced-motion:reduce){*{transition:none!important;animation:none!important}}
/* ---------- static figures: charts and bar comparisons ---------- */
.fig{padding:6px 6px 0;background-color:var(--stage);background-image:radial-gradient(var(--dot) 1px,transparent 1.2px);background-size:18px 18px}
.fig svg{display:block;width:100%;height:auto}
.gl{stroke:var(--bd);stroke-width:1;stroke-dasharray:3 5}.ax{stroke:var(--mut);stroke-width:1.4}
.tk{fill:var(--mut);font:500 11.5px 'JetBrains Mono',ui-monospace,monospace}.axl{fill:var(--mut);font:600 12px 'Vazirmatn',system-ui,sans-serif}
.cl-area{fill:url(#gf);opacity:0;transition:opacity 1s ease .9s}
.cl-line{fill:none;stroke:var(--acc);stroke-width:3.2;stroke-linecap:round;stroke-linejoin:round;stroke-dasharray:var(--len);stroke-dashoffset:var(--len);transition:stroke-dashoffset 1.5s cubic-bezier(.4,0,.2,1)}
.cl-pt{fill:var(--bg);stroke:var(--acc);stroke-width:3;opacity:0}.cl-lb{fill:var(--tx);font:700 11.5px 'JetBrains Mono',ui-monospace,monospace;opacity:0}
.run .cl-area{opacity:1}.run .cl-line{stroke-dashoffset:0}
.run .cl-pt{animation:fin .4s ease-out forwards;animation-delay:calc(var(--i)*.12s + .2s)}
.run .cl-lb{animation:fin .4s ease-out forwards;animation-delay:calc(var(--i)*.12s + .3s)}
@keyframes fin{from{opacity:0}to{opacity:1}}
.bars{display:flex;flex-direction:column;background:var(--bg)}
.brow{display:grid;grid-template-columns:minmax(110px,.8fr) minmax(0,2fr);gap:10px 18px;align-items:center;padding:14px 16px;border-bottom:1px solid var(--bd)}
.bname b{display:block;font:700 14px/1.4 'JetBrains Mono',ui-monospace,monospace;direction:ltr;text-align:right}
.bname span{display:block;font:500 11.5px/1.5 'Vazirmatn',system-ui,sans-serif;color:var(--mut)}
.bm{display:flex;flex-direction:column;gap:8px}
.m{display:grid;grid-template-columns:10.5rem minmax(24px,1fr) 5.2rem;gap:8px;align-items:center;font:600 12px/1.3 'Vazirmatn',system-ui,sans-serif;color:var(--mut);direction:ltr}
.m span{unicode-bidi:plaintext;text-align:left}
.mt{height:11px;border-radius:99px;background:var(--bd);overflow:hidden;direction:ltr}
.mt i{display:block;height:100%;width:0;border-radius:inherit;transition:width .9s cubic-bezier(.4,0,.2,1) .2s}
.run .mt i{width:var(--w)}
.m[data-k="0"] .mt i{background:var(--min-bd)}.m[data-k="1"] .mt i{background:var(--cmp-bd)}.m[data-k="2"] .mt i{background:var(--ok-bd)}
.m b{font:700 13px 'JetBrains Mono',ui-monospace,monospace;color:var(--tx);direction:ltr;text-align:left}
.cap{padding:12px 16px;border-top:1px solid var(--bd);font:500 13.5px/1.9 'Vazirmatn',system-ui,sans-serif;background:var(--bg)}
.sfoot{display:flex;align-items:center;gap:10px;padding:10px 14px;background:var(--panel);border-top:1px solid var(--bd);direction:rtl;font:500 12px 'Vazirmatn',system-ui,sans-serif;color:var(--mut)}
.card.nt *{transition:none!important;animation:none!important}
.card.narrow .brow{grid-template-columns:1fr}.card.narrow .m{grid-template-columns:6.4rem minmax(20px,1fr) 4.4rem;font-size:11px}
'''

_CF_JS = r'''(function(){
if(window.__CF&&window.__CF.v==="__VER__")return;
var CSS=__CSS__;
var IC={
 first:"M6 5h2v14H6zM19 5v14l-10-7z",last:"M16 5h2v14h-2zM5 5l10 7-10 7z",
 prev:"M17 5v14L6 12z",next:"M7 5v14l11-7z",
 play:"M8 5v14l11-7z",pause:"M6 5h4v14H6zM14 5h4v14h-4z",
 replay:"M12 4a8 8 0 1 0 7.7 10h-2.2A6 6 0 1 1 12 6v3.5L17 5l-5-4.5z",
 fs:"M4 9V4h5v2H6v3zM15 4h5v5h-2V6h-3zM4 15h2v3h3v2H4zM18 15h2v5h-5v-2h3z",
 fsx:"M9 4v5H4V7h3V4zM15 4h2v3h3v2h-5zM4 15h5v5H7v-3H4zM15 15h5v2h-3v3h-2z",
 loop:"M7 6h9V3l5 4.5-5 4.5V9H7a2 2 0 0 0-2 2v1H3v-1a4 4 0 0 1 4-5zm10 12H8v3l-5-4.5L8 12v3h9a2 2 0 0 0 2-2v-1h2v1a4 4 0 0 1-4 5z"
};
function svg(p){return '<svg viewBox="0 0 24 24" aria-hidden="true"><path d="'+p+'"/></svg>'}
function esc(s){return String(s).replace(/[&<>"]/g,function(c){return {"&":"&amp;","<":"&lt;",">":"&gt;",'"':"&quot;"}[c]})}
function isDark(){
 var d=document.documentElement,b=document.body||d;
 if(d.classList.contains("dark"))return true;
 if(d.classList.contains("light"))return false;
 var th=(d.getAttribute("data-theme")||"")+" "+((b.getAttribute&&b.getAttribute("data-vscode-theme-kind"))||"");
 if(/dark/i.test(th))return true;if(/light/i.test(th))return false;
 var bc=b.className||"";
 if(/vscode-light|vscode-high-contrast-light/.test(bc))return false;
 if(/vscode-dark|vscode-high-contrast/.test(bc))return true;
 if(b.getAttribute&&b.getAttribute("data-jp-theme-light")==="false")return true;
 if(b.getAttribute&&b.getAttribute("data-jp-theme-light")==="true")return false;
 return !!(window.matchMedia&&matchMedia("(prefers-color-scheme: dark)").matches);
}
function watchTheme(host,card){
 function t(){card.setAttribute("data-theme",isDark()?"dark":"light")}
 t();
 try{
  var mo=function(){if(host.isConnected)t()};
  new MutationObserver(mo).observe(document.documentElement,{attributes:true,attributeFilter:["class","data-theme"]});
  if(document.body)new MutationObserver(mo).observe(document.body,{attributes:true,attributeFilter:["class","data-vscode-theme-kind","data-jp-theme-light"]});
  if(window.matchMedia)matchMedia("(prefers-color-scheme: dark)").addEventListener("change",mo);
 }catch(e){}
}
function head(S,withFs){
 return '<div class="hd"><div class="ttl"><i class="dot"></i><span class="tt">'+esc(S.title)+'</span></div><div class="hr"><span class="bdg">'+esc(S.badge||"Interactive")+'</span>'+
  (withFs?'<button class="ib" data-a="fs" aria-label="تمام‌صفحه" title="تمام‌صفحه (F)">'+svg(IC.fs)+'</button>':'')+'</div></div>';
}
function mountScene(host,S){
 if(host.__cf)return;host.__cf=1;
 var root=host.attachShadow?host.attachShadow({mode:"open"}):host;
 var F=S.frames,N=F.length,n=S.items.length,isGrid=!!S.grid,LH=22;
 var codeH="";
 S.code.forEach(function(l,q){codeH+='<div class="ln"><span class="no">'+(q+1)+'</span><span class="tx">'+l+'</span></div>'});
 var lg="";S.legend.forEach(function(l){lg+='<span><i data-c="'+l[0]+'"></i>'+l[1]+'</span>'});
 function tb(a,ic,label,cls){return '<button class="tb '+(cls||"")+'" data-a="'+a+'" aria-label="'+label+'" title="'+label+'">'+svg(ic)+'</button>'}
 root.innerHTML='<style>'+CSS+'</style><div class="card nt" tabindex="0" role="group" aria-label="'+esc(S.title)+'">'+head(S,true)+
  '<div class="stage"></div><div class="lg">'+lg+'</div>'+
  '<div class="split"><div class="codep"><div class="chd"><s></s>'+esc(S.codeTitle)+'</div><div class="cbody"><div class="lines"><div class="hlb"></div>'+codeH+'</div></div></div>'+
  '<div class="info"><div><div class="bt">متغیرها</div><div class="vars"></div></div>'+
  '<div><div class="bt">این مرحله چه می‌کند؟</div><div class="note" aria-live="polite"></div></div>'+
  (S.showOut?'<div class="out"><div class="bt">خروجی برنامه</div><pre></pre></div>':'')+'</div></div>'+
  '<div class="dock"><div class="tp">'+tb("first",IC.first,"اولین گام")+tb("prev",IC.prev,"گام قبل")+tb("play",IC.play,"پخش","go")+tb("next",IC.next,"گام بعد")+tb("last",IC.last,"آخرین گام")+'</div>'+
  '<div class="tl"><div class="tinfo"><span class="tcap"></span><b class="cnt"></b></div><div class="track" role="slider" tabindex="0" aria-label="گام" aria-valuemin="1" aria-valuemax="'+N+'"><div class="rail"><div class="fill"></div></div><div class="ticks"></div><div class="thumb"></div></div></div>'+
  '<div class="sp" role="group" aria-label="سرعت"><button data-s="0.5">0.5×</button><button data-s="1" class="on">1×</button><button data-s="2">2×</button></div>'+
  '<button class="ib tg" data-a="loop" aria-label="تکرار خودکار" title="تکرار خودکار">'+svg(IC.loop)+'</button></div></div>';
 var $=function(s){return root.querySelector(s)};
 var card=$(".card"),stage=$(".stage"),hlb=$(".hlb"),cbody=$(".cbody"),lines=root.querySelectorAll(".ln"),
     varsEl=$(".vars"),noteEl=$(".note"),outPre=$(".out pre"),track=$(".track"),fill=$(".fill"),thumb=$(".thumb"),ticks=$(".ticks"),
     cnt=$(".cnt"),tcap=$(".tcap"),bPlay=$('[data-a=play]'),bLoop=$('[data-a=loop]');
 var st={k:0,playing:false,timer:null,loop:false,speed:1,last:-1,vars:"",pseudo:false};
 /* ---- stage ---- */
 var boxes=[],tags=[],rowH=[],colH=[],pe={};
 var sd=document.createElement("div");
 if(!isGrid){
  sd.className="row";
  for(var i=0;i<n;i++){
   var c=document.createElement("div");c.className="cell";
   var b=document.createElement("div");b.className="box";b.textContent=S.items[i];
   var ix=document.createElement("div");ix.className="idx";ix.textContent=(S.labels&&S.labels[i]!==undefined)?S.labels[i]:i;
   var tg=document.createElement("div");tg.className="tags";
   c.appendChild(b);c.appendChild(ix);c.appendChild(tg);sd.appendChild(c);boxes.push(b);tags.push(tg);
  }
 }else{
  var R=S.grid.r,C=S.grid.c;
  sd.className="grid";sd.style.gridTemplateColumns="auto repeat("+C+",var(--bw,52px))";
  sd.appendChild(document.createElement("div"));
  for(var j=0;j<C;j++){var h=document.createElement("div");h.className="hdr";h.textContent=S.grid.cl?S.grid.cl[j]:"j="+j;sd.appendChild(h);colH.push(h)}
  for(var r=0;r<R;r++){
   var rh=document.createElement("div");rh.className="hdr";rh.textContent=S.grid.rl?S.grid.rl[r]:"i="+r;sd.appendChild(rh);rowH.push(rh);
   for(var c2=0;c2<C;c2++){var bx=document.createElement("div");bx.className="box";bx.textContent=S.items[r*C+c2];sd.appendChild(bx);boxes.push(bx)}
  }
 }
 stage.appendChild(sd);
 function layout(){
  var W=stage.clientWidth||600,s,g,bw,fs,ml=1;
  S.items.forEach(function(t){if(String(t).length>ml)ml=String(t).length});
  if(!isGrid){
   var cw=Math.max(30,Math.min(150,Math.floor((W-36)/n/1.16)));
   s=Math.max(30,Math.min(72,cw));fs=Math.round(s*.34);
   var want=Math.max(s,Math.round(ml*fs*.62+22));
   bw=Math.min(want,cw);fs=Math.max(10,Math.min(fs,Math.floor((bw-14)/(ml*.62))));g=Math.round(Math.min(bw,72)*.2);
  }else{
   var C=S.grid.c;bw=Math.max(32,Math.min(60,Math.floor((W-120)/C*.86)));s=bw;g=8;fs=Math.max(10,Math.min(Math.round(s*.36),Math.floor((bw-10)/(ml*.62))));
  }
  card.style.setProperty("--s",s+"px");card.style.setProperty("--bw",bw+"px");card.style.setProperty("--fs",fs+"px");card.style.setProperty("--g",g+"px");
 }
 function sbTick(){
  ticks.innerHTML="";
  if(N<=60){for(var q=1;q<N-1;q++){var t=document.createElement("i");t.className="tick";t.style.left=(q/(N-1)*100)+"%";t.style.top="5px";ticks.appendChild(t)}}
 }
 ticks.style.cssText="position:absolute;left:0;right:0;top:0;height:22px;pointer-events:none";
 function render(k){
  var f=F[k];
  for(var i=0;i<boxes.length;i++){
   boxes[i].setAttribute("data-c",f.cls[i]||"i");
   var v=(f.vals&&f.vals[i]!==undefined)?f.vals[i]:S.items[i];
   if(boxes[i].textContent!==String(v))boxes[i].textContent=v;
  }
  if(!isGrid){
   var pr=f.ptr||{};
   Object.keys(pe).forEach(function(nm){if(pr[nm]===undefined||pr[nm]===null||pr[nm]<0||pr[nm]>=n){if(pe[nm].el.parentNode)pe[nm].el.parentNode.removeChild(pe[nm].el);delete pe[nm]}});
   Object.keys(pr).forEach(function(nm){
    var idx=pr[nm];if(idx===null||idx<0||idx>=n)return;
    if(pe[nm]&&pe[nm].idx===idx)return;
    if(pe[nm]&&pe[nm].el.parentNode)pe[nm].el.parentNode.removeChild(pe[nm].el);
    var el=document.createElement("span");el.className="ptr";el.textContent=nm;tags[idx].appendChild(el);pe[nm]={el:el,idx:idx};
   });
  }else{
   rowH.forEach(function(el,r){el.classList.toggle("on",f.row===r)});
   colH.forEach(function(el,c){el.classList.toggle("on",f.col===c)});
  }
  var a=f.ln[0],b=f.ln[1];
  hlb.style.transform="translateY("+((a-1)*LH)+"px)";hlb.style.height=((b-a+1)*LH)+"px";
  for(var q=0;q<lines.length;q++)lines[q].classList.toggle("on",q+1>=a&&q+1<=b);
  var topY=(a-1)*LH,botY=b*LH+16;
  if(topY<cbody.scrollTop+8||botY>cbody.scrollTop+cbody.clientHeight-8){cbody.scrollTop=Math.max(0,topY-cbody.clientHeight/2+LH*1.5)}
  var prev={};try{prev=JSON.parse(st.vars||"{}")}catch(e){}
  var h="",nv={};
  (f.vars||[]).forEach(function(v){nv[v[0]]=v[1];var ch=prev[v[0]]!==undefined&&prev[v[0]]!==v[1];h+='<div class="chip'+(v[1]==="—"?" na":"")+(ch?" chg":"")+'"><b>'+esc(v[0])+'</b><span>'+esc(v[1])+'</span></div>'});
  varsEl.innerHTML=h;st.vars=JSON.stringify(nv);
  if(st.last!==k){noteEl.innerHTML=f.note;noteEl.classList.remove("in");void noteEl.offsetWidth;noteEl.classList.add("in")}
  if(outPre){outPre.textContent=f.out||"";outPre.scrollTop=outPre.scrollHeight}
  var pct=N>1?k/(N-1)*100:100;
  fill.style.width=pct+"%";thumb.style.left=pct+"%";
  tcap.textContent=(k===N-1?"پایان اجرا":"گام");
  cnt.textContent=(k+1)+" / "+N;
  track.setAttribute("aria-valuenow",k+1);
  root.querySelector("[data-a=first]").disabled=root.querySelector("[data-a=prev]").disabled=(k===0);
  root.querySelector("[data-a=next]").disabled=root.querySelector("[data-a=last]").disabled=(k===N-1);
  bPlay.innerHTML=svg(st.playing?IC.pause:(k===N-1?IC.replay:IC.play));
  var lb=st.playing?"توقف":(k===N-1?"پخش دوباره":"پخش");bPlay.setAttribute("aria-label",lb);bPlay.title=lb;
  st.last=k;
 }
 function go(k){st.k=Math.max(0,Math.min(N-1,k));render(st.k)}
 function stop(){st.playing=false;clearTimeout(st.timer);render(st.k)}
 function delay(){return (S.interval||1200)/st.speed}
 function tick(){
  if(!st.playing)return;
  if(st.k<N-1)go(st.k+1);
  if(st.k===N-1){if(st.loop){st.timer=setTimeout(function(){go(0);st.timer=setTimeout(tick,delay())},delay());return}stop();return}
  st.timer=setTimeout(tick,delay());
 }
 function play(){if(st.k===N-1)go(0);st.playing=true;render(st.k);clearTimeout(st.timer);st.timer=setTimeout(tick,delay())}
 root.querySelector(".dock").addEventListener("click",function(e){
  var t=e.target.closest&&e.target.closest("[data-a],[data-s]");if(!t)return;
  if(t.hasAttribute("data-s")){st.speed=+t.getAttribute("data-s");root.querySelectorAll(".sp button").forEach(function(x){x.classList.toggle("on",x===t)});if(st.playing){clearTimeout(st.timer);st.timer=setTimeout(tick,delay())}return}
  var a=t.getAttribute("data-a");
  if(a==="play"){st.playing?stop():play()}
  else if(a==="loop"){st.loop=!st.loop;bLoop.classList.toggle("on",st.loop)}
  else{if(st.playing)stop();go(a==="first"?0:a==="last"?N-1:a==="prev"?st.k-1:st.k+1)}
 });
 function fromX(e){var r=track.getBoundingClientRect();var p=r.width?(e.clientX-r.left)/r.width:0;return Math.round(Math.max(0,Math.min(1,p))*(N-1))}
 var drag=false;
 track.addEventListener("pointerdown",function(e){drag=true;try{track.setPointerCapture(e.pointerId)}catch(_){}if(st.playing)stop();go(fromX(e))});
 track.addEventListener("pointermove",function(e){if(drag)go(fromX(e))});
 track.addEventListener("pointerup",function(){drag=false});
 track.addEventListener("pointercancel",function(){drag=false});
 track.addEventListener("keydown",function(e){var k=e.key;
  if(k==="ArrowRight"||k==="ArrowUp"){e.preventDefault();e.stopPropagation();if(st.playing)stop();go(st.k+1)}
  else if(k==="ArrowLeft"||k==="ArrowDown"){e.preventDefault();e.stopPropagation();if(st.playing)stop();go(st.k-1)}});
 card.addEventListener("keydown",function(e){
  var k=e.key,tn=e.target&&e.target.tagName;
  if(k===" "||k==="Spacebar"){if(tn==="BUTTON")return;e.preventDefault();st.playing?stop():play()}
  else if(k==="ArrowRight"){e.preventDefault();if(st.playing)stop();go(st.k+1)}
  else if(k==="ArrowLeft"){e.preventDefault();if(st.playing)stop();go(st.k-1)}
  else if(k==="Home"){e.preventDefault();go(0)}else if(k==="End"){e.preventDefault();go(N-1)}
  else if(k==="f"||k==="F"){toggleFs()}
 });
 var bFs=$("[data-a=fs]");
 function fsOn(){return document.fullscreenElement===host||st.pseudo}
 function syncFs(){var on=fsOn();card.classList.toggle("full",on);if(bFs){bFs.innerHTML=svg(on?IC.fsx:IC.fs)}setTimeout(relayout,60);setTimeout(relayout,300)}
 function toggleFs(){
  if(fsOn()){if(st.pseudo){st.pseudo=false;syncFs()}else if(document.exitFullscreen)document.exitFullscreen();return}
  var fb=function(){st.pseudo=true;syncFs()};
  try{if(host.requestFullscreen){var p=host.requestFullscreen();if(p&&p.catch)p.catch(fb)}else fb()}catch(_){fb()}
 }
 if(bFs)bFs.addEventListener("click",toggleFs);
 document.addEventListener("fullscreenchange",function(){if(host.isConnected)syncFs()});
 document.addEventListener("keydown",function(e){if(e.key==="Escape"&&st.pseudo){st.pseudo=false;syncFs()}});
 function relayout(){
  card.classList.toggle("narrow",(card.clientWidth||700)<640);
  card.classList.add("nt");layout();render(st.k);
  requestAnimationFrame(function(){requestAnimationFrame(function(){card.classList.remove("nt")})});
 }
 watchTheme(host,card);
 sbTick();
 if(window.ResizeObserver){new ResizeObserver(function(){if(host.isConnected)relayout()}).observe(card)}
 else window.addEventListener("resize",relayout);
 relayout();
}
function mountStatic(host,S){
 if(host.__cf)return;host.__cf=1;
 var root=host.attachShadow?host.attachShadow({mode:"open"}):host;
 root.innerHTML='<style>'+CSS+'</style><div class="card nt" role="group" aria-label="'+esc(S.title)+'">'+head(S,false)+'<div class="'+(S.body||"fig")+'">'+S.html+'</div>'+
  (S.caption?'<div class="cap">'+S.caption+'</div>':'')+
  '<div class="sfoot"><button class="ib" data-a="replay" aria-label="پخش دوباره" title="پخش دوباره">'+svg(IC.replay)+'</button><span>برای دیدن دوباره‌ی انیمیشن، دکمه‌ی پخش دوباره را بزنید</span></div></div>';
 var card=root.querySelector(".card");
 watchTheme(host,card);
 function nar(){card.classList.toggle("narrow",(card.clientWidth||700)<560)}
 nar();if(window.ResizeObserver){new ResizeObserver(function(){if(host.isConnected)nar()}).observe(card)}
 function start(){card.classList.remove("run");requestAnimationFrame(function(){requestAnimationFrame(function(){card.classList.remove("nt");card.classList.add("run")})})}
 root.querySelector("[data-a=replay]").addEventListener("click",function(){card.classList.remove("run");void card.offsetWidth;requestAnimationFrame(function(){card.classList.add("run")})});
 start();
}
window.__CF={v:"__VER__",mount:function(h,S){return S.kind==="static"?mountStatic(h,S):mountScene(h,S)}};
})();
'''

_KW = {"if", "else", "for", "while", "do", "return", "break", "continue", "switch", "case",
       "default", "goto", "sizeof", "struct", "typedef", "const", "static"}
_TY = {"int", "char", "void", "float", "double", "long", "short", "unsigned", "signed", "size_t"}
_TOK = re.compile(r'(//.*$)|(/\*.*?\*/)|("(?:\\.|[^"\\])*")|(\'(?:\\.|[^\'\\])*\')|(#\s*\w+)|(\b\d+(?:\.\d+)?\b)|(\b[A-Za-z_]\w*\b)')


def _hl(line):
    out, pos = [], 0
    for m in _TOK.finditer(line):
        out.append(_html.escape(line[pos:m.start()]))
        t = m.group(0)
        if m.group(1) or m.group(2): cls = "c"
        elif m.group(3) or m.group(4): cls = "s"
        elif m.group(5): cls = "p"
        elif m.group(6): cls = "n"
        else:
            after = line[m.end():].lstrip()[:1]
            if t in _KW: cls = "k"
            elif t in _TY: cls = "t"
            elif after == "(": cls = "f"
            else: cls = None
        e = _html.escape(t)
        out.append('<span class="%s">%s</span>' % (cls, e) if cls else e)
        pos = m.end()
    out.append(_html.escape(line[pos:]))
    return "".join(out)


def _c(x):
    """inline code span used inside Persian notes"""
    return '<code dir="ltr">%s</code>' % _html.escape(str(x))


def _widget(scene):
    sid = "cf" + uuid.uuid4().hex[:8]
    ver = hashlib.md5((_CF_CSS + _CF_JS).encode("utf8")).hexdigest()[:10]
    lib = _CF_JS.replace("__CSS__", json.dumps(_CF_CSS)).replace("__VER__", ver)
    data = json.dumps(scene, ensure_ascii=False, separators=(",", ":")).replace("</", "<\\/")
    return HTML(
        '<div data-cf="1" id="%s" style="display:block"></div>'
        '<script>%s\n(function(){var S=%s;var id="%s";var tries=0;'
        '(function go(){var host=document.getElementById(id);'
        'if(host&&host.isConnected){window.__CF.mount(host,S)}else if(tries++<200){setTimeout(go,50)}})();})();</script>'
        % (sid, lib, data, sid))


class Rec:
    """Record an execution step by step and render it as an interactive player."""
    def __init__(self, title, code, items, legend, labels=None, grid=None,
                 code_title="C · main.c", show_out=None, interval=1100):
        self.title, self.code, self.items = title, code, [str(i) for i in items]
        self.legend, self.labels, self.grid = legend, labels, grid
        self.code_title, self.show_out, self.interval = code_title, show_out, interval
        self.c = ["i"] * len(items)
        self.p, self.v, self.o, self.frames = {}, {}, "", []

    def var(self, **kw):
        self.v.update(kw)

    def mark(self, spec, cls=None):
        if isinstance(spec, dict):
            for k, v in spec.items(): self.c[k] = v
        else:
            for k in ([spec] if isinstance(spec, int) else spec): self.c[k] = cls

    def fill(self, cls):
        self.c = [cls] * len(self.c)

    def ptr(self, **kw):
        for k, v in kw.items():
            if v is None: self.p.pop(k, None)
            else: self.p[k] = v

    def echo(self, text):
        self.o += text

    def at(self, ln, note, row=None, col=None, vals=None):
        if isinstance(ln, int): ln = [ln, ln]
        f = dict(ln=list(ln), cls=list(self.c), ptr=dict(self.p),
                 vars=[[k, str(v)] for k, v in self.v.items()], note=note, out=self.o)
        if row is not None: f["row"] = row
        if col is not None: f["col"] = col
        if vals is not None: f["vals"] = [str(x) for x in vals]
        self.frames.append(f)

    def show(self):
        s = dict(title=self.title, badge="Step by step", codeTitle=self.code_title,
                 code=[_hl(l) for l in self.code], items=self.items, frames=self.frames,
                 legend=self.legend, interval=self.interval,
                 showOut=(any(f['out'] for f in self.frames) if self.show_out is None else self.show_out))
        if self.labels is not None: s["labels"] = [str(x) for x in self.labels]
        if self.grid: s["grid"] = self.grid
        return _widget(s)


def _static(title, html, caption="", badge="Figure", body="fig"):
    return _widget(dict(kind="static", title=title, badge=badge, html=html, caption=caption, body=body))


def fig_line(title, points, xlabel, ylabel, ymax, yticks, caption="", label_points=True):
    W, H, ml, mr, mt, mb = 720, 330, 62, 24, 34, 54
    iw, ih = W - ml - mr, H - mt - mb
    xmax = points[-1][0]
    X = lambda x: ml + x / xmax * iw
    Y = lambda y: mt + ih - y / ymax * ih
    p = ['<svg viewBox="0 0 %d %d" role="img" aria-label="%s">' % (W, H, _html.escape(title)),
         '<defs><linearGradient id="gf" x1="0" y1="0" x2="0" y2="1"><stop offset="0" style="stop-color:var(--acc);stop-opacity:.28"/>'
         '<stop offset="1" style="stop-color:var(--acc);stop-opacity:0"/></linearGradient></defs>']
    for t in yticks:
        p.append('<line class="gl" x1="%d" x2="%d" y1="%.1f" y2="%.1f"/><text class="tk" x="%d" y="%.1f" text-anchor="end">%s</text>' % (ml, W - mr, Y(t), Y(t), ml - 10, Y(t) + 4, t))
    for x, _ in points:
        p.append('<text class="tk" x="%.1f" y="%d" text-anchor="middle">%s</text>' % (X(x), H - mb + 20, x))
    p.append('<line class="ax" x1="%d" x2="%d" y1="%.1f" y2="%.1f"/>' % (ml, W - mr, Y(0), Y(0)))
    d, ln, px, py = "", 0.0, None, None
    for k, (x, y) in enumerate(points):
        a, b = X(x), Y(y)
        d += ("L" if k else "M") + "%.1f %.1f" % (a, b)
        if k: ln += ((a - px) ** 2 + (b - py) ** 2) ** .5
        px, py = a, b
    p.append('<path class="cl-area" d="%s L%.1f %.1f L%.1f %.1f Z"/>' % (d, X(xmax), Y(0), X(0), Y(0)))
    p.append('<path class="cl-line" d="%s" style="--len:%d"/>' % (d, int(ln) + 1))
    for k, (x, y) in enumerate(points):
        if label_points:
            p.append('<text class="cl-lb" style="--i:%d" x="%.1f" y="%.1f" text-anchor="middle">%s</text>' % (k, X(x), Y(y) - 12, y))
        p.append('<circle class="cl-pt" style="--i:%d" cx="%.1f" cy="%.1f" r="5.5"><title>%s: %s</title></circle>' % (k, X(x), Y(y), x, y))
    p.append('<text class="axl" x="%.1f" y="%d" text-anchor="middle">%s</text>' % (ml + iw / 2, H - 8, _html.escape(xlabel)))
    p.append('<text class="axl" transform="translate(14 %.1f) rotate(-90)" text-anchor="middle">%s</text></svg>' % (mt + ih / 2, _html.escape(ylabel)))
    return _static(title, "".join(p), caption, "Chart")


def fig_bars(title, rows, caption="", metric_labels=None):
    """rows: [dict(name, sub, metrics=[(label, value, max)])]"""
    h = []
    for r in rows:
        h.append('<div class="brow"><div class="bname"><b>%s</b><span>%s</span></div><div class="bm">' % (_html.escape(r["name"]), _html.escape(r.get("sub", ""))))
        for k, (lab, val, mx) in enumerate(r["metrics"]):
            w = max(val / mx * 100, 3) if val else 0
            h.append('<div class="m" data-k="%d"><span>%s</span><div class="mt"><i style="--w:%.2f%%"></i></div><b>%s</b></div>' % (k, _html.escape(lab), w, val))
        h.append('</div></div>')
    return _static(title, "".join(h), caption, "Comparison", body="bars")


# ============================================================
# صحنه‌های انیمیشن بخش سوم (هر تابع، یک پخش‌کننده‌ی تعاملی نمایش می‌دهد)
# ============================================================
import math

_LG_PATH = [["cmp", "شرط در حال بررسی"], ["dim", "نادرست بود؛ رد شد"], ["ok", "شرطِ برنده"], ["i", "هنوز بررسی نشده"]]
_LG_SW = [["cur", "در حال اجرا"], ["ok", "اجرا شد"], ["dim", "از روی آن پریده شد"], ["i", "هنوز اجرا نشده"]]
_LG_LOOP = [["cur", "تکرار جاری"], ["vis", "تکرار انجام‌شده"], ["min", "شرط نادرست شد"], ["i", "هنوز نرسیده‌ایم"]]


def show_leap(year):
    code = ["if (year % 400 == 0)", "    leap = 1;", "else if (year % 100 == 0)", "    leap = 0;",
            "else", "    leap = (year % 4 == 0);"]
    r = Rec(f"Leap year: decision path for year = {year}", code, ["year % 400", "year % 100", "else"],
            _LG_PATH, labels=["شرط ۱", "شرط ۲", "شرط ۳"])
    r.var(year=year, leap="—")
    r.at(1, f"{_c('year')} برابر {_c(year)} است و {_c('leap')} هنوز مقدار ندارد. شرط‌ها از بالا به پایین بررسی می‌شوند.")
    for k, (d, val) in enumerate([(400, 1), (100, 0)]):
        if k: r.mark(k - 1, "dim")
        r.mark(k, "cmp"); rem = year % d
        res = "<b>درست</b> است." if rem == 0 else "نادرست است؛ سراغ شرط بعدی می‌رویم."
        r.at(1 if k == 0 else 3, f"باقی‌مانده‌ی {_c(f'{year} % {d}')} برابر {_c(rem)} است ← شرط {res}")
        if rem == 0:
            r.mark(k, "ok"); r.var(leap=val)
            r.at(2 if k == 0 else 4, f"این شاخه برنده شد و {_c('leap')} می‌شود {_c(val)}. بقیه‌ی شاخه‌ها بدون بررسی رد می‌شوند.")
            r.mark({j: "dim" for j in range(k + 1, 3)}); r.at([1, 6], "پایان تصمیم.")
            return display(r.show())
    r.mark(1, "dim"); r.mark(2, "ok")
    r.at(5, "هیچ‌کدام از دو شرط بالا برقرار نبود؛ شاخه‌ی else اجرا می‌شود.")
    val = int(year % 4 == 0); r.var(leap=val)
    r.at(6, f"{_c('year % 4')} برابر {_c(year % 4)} است، پس {_c('leap')} = {_c(val)}.")
    display(r.show())


def show_ladder(score):
    code = ["if (score < 0 || score > 20)", "    printf(\"invalid\");", "else if (score >= 17)", "    printf(\"Excellent\");",
            "else if (score >= 14)", "    printf(\"Good\");", "else if (score >= 10)", "    printf(\"Pass\");",
            "else", "    printf(\"Fail\");"]
    r = Rec(f"else-if ladder: score = {score}", code, ["invalid", "≥ 17", "≥ 14", "≥ 10", "else"], _LG_PATH,
            labels=[f"شاخه {x}" for x in "۱۲۳۴۵"], interval=1300)
    r.var(score=score)
    r.at(1, f"نمره‌ی {_c(score)} وارد زنجیره می‌شود. «اولین شرطِ درست برنده است» و بقیه‌ی زنجیره بررسی نمی‌شود.")
    tests = [(0, 1, "score < 0 || score > 20", score < 0 or score > 20, "invalid"), (1, 3, "score >= 17", score >= 17, "Excellent"),
             (2, 5, "score >= 14", score >= 14, "Good"), (3, 7, "score >= 10", score >= 10, "Pass")]
    for k, ln, ex, res, word in tests:
        if k: r.mark(k - 1, "dim")
        r.mark(k, "cmp")
        r.at(ln, f"{_c(ex)} با {_c(f'score = {score}')} ← " + ("<b>درست</b>؛ این شاخه برنده است." if res else "نادرست؛ به شرط بعدی می‌رویم."))
        if res:
            r.mark(k, "ok"); r.mark({j: "dim" for j in range(k + 1, 5)}); r.echo(word)
            r.at(ln + 1, f"پیام {_c(word)} چاپ شد. شاخه‌های بعدی (حتی اگر درست بودند) دیگر بررسی نمی‌شوند.")
            return display(r.show())
    r.mark(3, "dim"); r.mark(4, "ok"); r.echo("Fail")
    r.at(10, "هیچ شرطی برقرار نبود؛ else یعنی «بقیه‌ی حالت‌ها».")
    display(r.show())


def show_switch(with_break):
    if with_break:
        code = ["switch (level) {", "    case 3: printf(\"A \"); break;", "    case 2: printf(\"B \"); break;",
                "    case 1: printf(\"C \"); break;", "    default: printf(\"D \");", "}"]
    else:
        code = ["switch (level) {", "    case 3: printf(\"A \");", "    case 2: printf(\"B \");",
                "    case 1: printf(\"C \");", "    default: printf(\"D \");", "}"]
    r = Rec("switch " + ("WITH" if with_break else "WITHOUT") + " break (level = 2)", code,
            ["case 3", "case 2", "case 1", "default"], _LG_SW, labels=["مقدار ۳", "مقدار ۲", "مقدار ۱", "هر مقدار دیگر"])
    r.var(level=2)
    r.at(1, f"{_c('switch')} مقدار {_c('level')} را فقط یک‌بار حساب می‌کند: {_c(2)}.")
    r.mark(0, "dim"); r.mark(1, "cur")
    r.at(3, f"پرش مستقیم به {_c('case 2')}. «پرش» فقط نقطه‌ی <b>شروع</b> اجراست.")
    r.echo("B "); r.mark(1, "ok")
    if with_break:
        r.at(3, f"{_c('B')} چاپ شد.")
        r.mark({2: "dim", 3: "dim"})
        r.at(3, f"{_c('break')}: کنترل فوراً از کل {_c('switch')} خارج می‌شود.")
        r.at(6, "پایان switch. خروجی فقط «B» است.")
    else:
        r.at(3, f"{_c('B')} چاپ شد. {_c('break')} نداریم، پس برنامه متوقف نمی‌شود.")
        r.mark(2, "cur"); r.echo("C ")
        r.at(4, f"سقوط (Fall-Through) به {_c('case 1')}: بدون بررسی شرط، {_c('C')} هم چاپ می‌شود.")
        r.mark(2, "ok"); r.mark(3, "cur"); r.echo("D ")
        r.at(5, f"سقوط به {_c('default')}: {_c('D')} هم چاپ می‌شود.")
        r.mark(3, "ok"); r.at(6, "پایان switch. خروجی «B C D» است: فراموش کردن break سه بلوک را اجرا کرد.")
    display(r.show())


def show_digit_sum(n=472):
    code = ["int sum = 0;", "while (n > 0) {", "    sum += n % 10;", "    n /= 10;", "}"]
    digs = [int(x) for x in str(n)]
    r = Rec(f"while loop: digit sum of {n}", code, digs,
            [["cur", "رقمی که الان جدا می‌شود"], ["dim", "از عدد حذف شد"], ["i", "هنوز در عدد است"]],
            labels=["صدگان", "دهگان", "یکان"][-len(digs):])
    s, idx, m = 0, len(digs) - 1, n
    r.var(n=m, sum=0)
    r.at(1, f"{_c('sum')} را صفر می‌کنیم. ارقام را از «یکان» به سمت چپ برمی‌داریم.")
    while m > 0:
        r.at(2, f"شرط {_c('n > 0')} با {_c(f'n = {m}')} درست است.")
        r.mark(idx, "cur"); r.ptr(**{"n%10": idx}); s += m % 10; r.var(sum=s)
        r.at(3, f"{_c('n % 10')} برابر {_c(m % 10)} (رقم آخر) است و به {_c('sum')} اضافه می‌شود: {_c(s)}.")
        m //= 10; r.var(n=m); r.mark(idx, "dim"); idx -= 1
        r.at(4, f"{_c('n /= 10')} ← رقم آخر حذف شد و {_c('n')} می‌شود {_c(m)}.")
    r.ptr(**{"n%10": None})
    r.at(2, f"{_c('0 > 0')} نادرست است؛ حلقه تمام شد. مجموع ارقام {_c(s)} است.")
    display(r.show())


def show_for_order():
    code = ["for (i = 0; i < 3; i++) {", "    printf(\"%d \", i);", "}"]
    r = Rec("for loop: the real order of execution", code, [0, 1, 2, 3], _LG_LOOP, labels=["i=0", "i=1", "i=2", "i=3"])
    r.var(i="—")
    r.at(1, "سه بخش داخل پرانتز به ترتیب خاصی اجرا می‌شوند: مقداردهی، شرط، (بدنه)، به‌روزرسانی.")
    r.var(i=0); r.mark(0, "cur"); r.ptr(i=0)
    r.at(1, f"① <b>مقداردهی</b>: {_c('i = 0')}. فقط یک‌بار و در ابتدا اجرا می‌شود.")
    for i in range(3):
        r.at(1, f"② <b>شرط</b>: {_c(f'{i} < 3')} درست است، پس بدنه اجرا می‌شود.")
        r.echo(f"{i} "); r.at(2, f"③ <b>بدنه</b>: مقدار {_c(i)} چاپ می‌شود.")
        r.mark(i, "vis"); r.mark(i + 1, "cur"); r.ptr(i=i + 1); r.var(i=i + 1)
        r.at(1, f"④ <b>به‌روزرسانی</b>: {_c('i++')} ← {_c('i')} می‌شود {_c(i + 1)}. این بخش بعد از بدنه اجرا می‌شود، هرچند در خط اول نوشته شده.")
    r.mark(3, "min")
    r.at(1, f"② <b>شرط</b>: {_c('3 < 3')} نادرست است؛ خروج. بدنه ۳ بار اجرا شد، اما شرط ۴ بار بررسی شد.")
    display(r.show())


def show_euclid(a=48, b=18):
    code = ["while (b != 0) {", "    r = a % b;", "    a = b;", "    b = r;", "}", "printf(\"%d\", a);"]
    chain = [a, b]
    while chain[-1] != 0: chain.append(chain[-2] % chain[-1])
    r = Rec(f"Euclid's algorithm: gcd({a}, {b})", code, chain,
            [["cmp", "a (مقسوم)"], ["key", "b (مقسوم‌علیه)"], ["ok", "باقی‌مانده‌ی جدید"], ["vis", "دیگر استفاده نمی‌شود"], ["i", "هنوز محاسبه نشده"]],
            labels=["a0", "b0"] + [f"r{k}" for k in range(1, len(chain) - 1)])
    known = [2]
    vals = lambda: [str(chain[i]) if i < known[0] else "?" for i in range(len(chain))]
    r.var(a=a, b=b, r="—"); r.mark({0: "cmp", 1: "key"}); r.ptr(a=0, b=1)
    r.at(1, f"شروع با {_c(f'a = {a}')} و {_c(f'b = {b}')}. هر باقی‌مانده‌ی بعدی را خودِ حلقه می‌سازد.", vals=vals())
    k, x, y = 0, a, b
    while y != 0:
        r.at(1, f"{_c('b != 0')} ← {_c(f'{y} != 0')} درست است؛ وارد بدنه می‌شویم.", vals=vals())
        rem = x % y; known[0] += 1; r.var(r=rem); r.mark(k + 2, "ok")
        r.at(2, f"{_c('r = a % b')} ← {_c(f'{x} % {y}')} = {_c(rem)}. باقی‌مانده همیشه از {_c('b')} کوچک‌تر است؛ پس دنباله اکیداً نزولی است.", vals=vals())
        x = y; r.var(a=x); r.mark(k, "vis"); r.mark(k + 1, "cmp"); r.ptr(a=k + 1)
        r.at(3, f"{_c('a = b')} ← {_c('a')} می‌شود {_c(x)}.", vals=vals())
        y = rem; r.var(b=y); r.mark(k + 2, "key"); r.ptr(b=k + 2)
        r.at(4, f"{_c('b = r')} ← {_c('b')} می‌شود {_c(y)}.", vals=vals())
        k += 1
    r.at(1, f"{_c('0 != 0')} نادرست است؛ حلقه تمام شد. آخرین {_c('a')} همان ب.م.م است.", vals=vals())
    r.echo(str(x)); r.mark(k, "ok")
    r.at(6, f"gcd({a}, {b}) = {_c(x)}. ({k} بار بدنه اجرا شد.)", vals=vals())
    display(r.show())


def show_triangle(R=4):
    code = [f"for (int i = 1; i <= {R}; i++) {{", "    for (int j = 1; j <= i; j++) {", "        printf(\"*\");", "    }",
            "    printf(\"\\n\");", "}"]
    r = Rec(f"Nested loops: right triangle (n = {R})", code, ["·"] * (R * R),
            [["cur", "تکرار جاری"], ["vis", "ستاره چاپ شد"], ["dim", "خارج از j ≤ i"], ["i", "هنوز نرسیده‌ایم"]],
            grid=dict(r=R, c=R, rl=[f"i={x}" for x in range(1, R + 1)], cl=[f"j={x}" for x in range(1, R + 1)]), interval=700)
    shown, cnt = ["·"] * (R * R), 0
    r.var(i="—", j="—", iterations=0)
    r.at(1, f"سطر ← حلقه‌ی بیرونی ({_c('i')}) و ستون ← حلقه‌ی داخلی ({_c('j')}). فقط خانه‌هایی که {_c('j <= i')} هستند ستاره می‌گیرند.", vals=shown)
    for i in range(1, R + 1):
        for j in range(i + 1, R + 1): r.mark((i - 1) * R + j - 1, "dim")
        r.var(i=i, j="—")
        r.at(2, f"حلقه‌ی بیرونی: {_c(f'i = {i}')}. حلقه‌ی داخلی برای این سطر {i} بار اجرا می‌شود.", row=i - 1, vals=shown)
        for j in range(1, i + 1):
            idx = (i - 1) * R + j - 1; cnt += 1
            r.mark(idx, "cur"); shown[idx] = "*"; r.var(j=j, iterations=cnt); r.echo("*")
            r.at(3, f"{_c(f'i = {i}, j = {j}')}: شرط {_c('j <= i')} درست است و یک ستاره چاپ می‌شود. تکرارها تا الان: {_c(cnt)}.", row=i - 1, col=j - 1, vals=shown)
            r.mark(idx, "vis")
        r.echo("\n")
        r.at(5, f"{_c(f'j = {i + 1}')} شرط حلقه‌ی داخلی را نادرست کرد. خط جدید بیرون از حلقه‌ی داخلی چاپ می‌شود و {_c('i++')} اجرا می‌شود.", row=i - 1, vals=shown)
    r.at(6, f"پایان: بدنه‌ی درونی {_c(cnt)} بار اجرا شد؛ یعنی {_c('1+2+3+4 = 10')}، نه {_c('4×4 = 16')}.", vals=shown)
    display(r.show())


def show_break_continue():
    code = ["for (i = 1; i <= 8; i++) {", "    if (i % 3 == 0) continue;", "    if (i == 7) break;",
            "    sum += i;", "}", "printf(\"%d\", sum);"]
    r = Rec("break and continue inside one loop", code, list(range(1, 9)),
            [["cur", "تکرار جاری"], ["ok", "به sum اضافه شد"], ["dim", "continue: این دور رها شد"], ["min", "break: حلقه قطع شد"], ["i", "هنوز اجرا نشده"]],
            labels=[f"i={x}" for x in range(1, 9)])
    s = 0; r.var(i=1, sum=0)
    for i in range(1, 9):
        r.var(i=i); r.mark(i - 1, "cur"); r.ptr(i=i - 1)
        r.at(1, f"شروع تکرار با {_c(f'i = {i}')}.")
        if i % 3 == 0:
            r.mark(i - 1, "dim")
            r.at(2, f"{_c(f'{i} % 3 == 0')} درست است ← {_c('continue')}: بقیه‌ی بدنه رد می‌شود و کنترل به <b>به‌روزرسانی</b> ({_c('i++')}) می‌رود.")
            continue
        r.at(2, f"{_c(f'{i} % 3 == 0')} نادرست است؛ ادامه می‌دهیم.")
        if i == 7:
            r.mark(i - 1, "min")
            r.at(3, f"{_c('i == 7')} درست است ← {_c('break')}: کل حلقه فوراً قطع می‌شود، هرچند {_c('i <= 8')} هنوز برقرار است.")
            break
        r.at(3, f"{_c(f'{i} == 7')} نادرست است؛ ادامه می‌دهیم.")
        s += i; r.var(sum=s); r.mark(i - 1, "ok")
        r.at(4, f"{_c(f'sum += {i}')} ← {_c('sum')} می‌شود {_c(s)}.")
    r.echo(str(s))
    r.at(6, f"پایان. مجموع {_c(s)} است: ۱، ۲، ۴ و ۵ اضافه شدند؛ ۳ و ۶ با {_c('continue')} رد شدند و ۷ و ۸ هرگز به {_c('sum')} نرسیدند.")
    display(r.show())


def fig_primes():
    def nxt(m):
        while True:
            if m > 1 and all(m % d for d in range(2, math.isqrt(m) + 1)): return m
            m += 1
    rows = []
    for t in [100, 1000, 10000, 100000, 1000000]:
        p = nxt(t); slow, fast = p - 2, math.isqrt(p) - 1
        rows.append(dict(name=f"n = {p}", sub="عدد اول؛ بدترین حالت",
                         metrics=[("d = 2 … n-1   O(n)", slow, slow), ("d·d ≤ n   O(√n)", fast, slow)]))
    display(fig_bars("Primality test: number of loop iterations", rows,
                     "طول هر نوار نسبت به نسخه‌ی کند همان ردیف سنجیده شده است. هرچه n بزرگ‌تر شود، نسخه‌ی √n نسبتاً کوچک‌تر می‌شود: برای حدود یک میلیون، فقط حدود <code dir='ltr'>1000</code> گام به‌جای <code dir='ltr'>1000000</code> گام."))


def fig_growth():
    rows = []
    for n in (10, 100, 1000):
        rows.append(dict(name=f"n = {n}", sub="تعداد تکرار بدنه",
                         metrics=[("halving loop   O(log n)", int(math.log2(n)), n * n), ("single loop   O(n)", n, n * n),
                                  ("nested loops   O(n²)", n * n, n * n)]))
    display(fig_bars("How the number of iterations grows with n", rows,
                     "در هر ردیف طول نوارها نسبت به <code dir='ltr'>n²</code> سنجیده شده است. با ۱۰۰ برابر شدن <code dir='ltr'>n</code>، حلقه‌ی ساده ۱۰۰ برابر و حلقه‌ی تودرتو ۱۰۰۰۰ برابر کار می‌کند؛ اما حلقه‌ی نصف‌شونده فقط چند گام بیشتر."))


print("Visualization toolkit ready.")
````

<h2>مقدمه</h2>

<p>تا اینجای درس، همه‌ی برنامه‌های ما یک ویژگی مشترک داشتند: دستورها از اولین خط تا آخرین خط، <strong>دقیقاً یک‌بار</strong> و به همان ترتیبی که نوشته شده‌اند اجرا می‌شدند. به این الگو <strong>اجرای ترتیبی (Sequence)</strong> می‌گوییم. اما مسئله‌های واقعی این‌طور نیستند. یک سیستم بانکی باید ابتدا تصمیم بگیرد: «اگر موجودی کافی است، برداشت را ثبت کن؛ وگرنه پیام خطا بده». برنامه‌ی میانگین‌گیری باید یک کار را <strong>برای هر دانشجو</strong> تکرار کند، بی‌آن‌که تعداد دانشجوها در زمان نوشتن برنامه معلوم باشد. یعنی دو نیاز تازه داریم: <strong>تصمیم‌گیری</strong> و <strong>تکرار</strong>.</p>

<p>در بخش اول، الگوریتم‌هایی دیدید که از «اگر» و «تکرار کن» ساخته می‌شدند (مثل بازی حدس عدد) و کدشان را بدون توضیح سینتکس خواندید. حالا وقت آن است که همان ابزارها را دقیق یاد بگیریم. در <strong>فصل ۶</strong> ساختارهای شرطی (<code>if</code>، <code>else if</code>، <code>else</code>، عمل‌گر <code>?:</code> و <code>switch</code>) و در <strong>فصل ۷</strong> حلقه‌ها (<code>while</code>، <code>do-while</code>، <code>for</code>) و ابزارهای کنترل آن‌ها (<code>break</code>، <code>continue</code>) را می‌بینیم و در پایان، دلیل پرهیز از <code>goto</code> را بررسی می‌کنیم.</p>

<p>یک نکته درباره‌ی هدف این بخش: <strong>هدف ما یادگیری سینتکس نیست؛ هدف ما یادگیری الگوریتم‌هاست.</strong> سینتکس <code>if</code> و <code>for</code> را می‌توان در ده دقیقه یاد گرفت. چیزی که ارزشمند است، توانایی تبدیل یک مسئله به «تصمیم‌ها و تکرارهای دقیق» و اطمینان از درستی آن است. برای همین در طول بخش، برای هر مسئله چهار ابزار را کنار هم می‌بینید:</p>

<ul>
<li><strong>Pseudocode</strong> و <strong>Flowchart</strong> (که در بخش اول دیدید) برای طراحی الگوریتم.</li>
<li><strong>Trace (دنبال کردن دستی):</strong> نوشتن مقدار متغیرها در هر گام، مثل یک کامپیوتر.</li>
<li><strong>Test Cases:</strong> انتخاب ورودی‌هایی که همه‌ی مسیرها و لبه‌ها (Edge Cases) را آزمایش کنند؛ همان نگاهِ «ورودی صفر، منفی و بزرگ» که در بخش اول گفتیم.</li>
</ul>

<p>در طول این بخش یک سناریوی واحد داریم: <strong>برنامه‌ی ارزیابی نمرات کلاس</strong>؛ نمره‌ها را اعتبارسنجی، درجه‌بندی و جمع‌بندی می‌کنیم و در پایان یک منوی تعاملی می‌سازیم (که در بخش پنجم با آرایه‌ها کامل می‌شود). در کنار آن چند الگوریتم کلاسیک کوچک (سال کبیسه، شمارش ارقام، عدد اول، ب.م.م) می‌بینیم.</p>

<h4>۱) قوانین دمورگان (De Morgan's Laws)</h4>

<p>وقتی می‌خواهیم حالت‌های <strong>نامعتبر</strong> را از روی حالت‌های <strong>معتبر</strong> بسازیم، نقیض شرط را می‌نویسیم:</p>

<table>
<thead>
<tr><th>عبارت اصلی</th><th>عبارت معادل</th></tr>
</thead>
<tbody>
<tr><td><code>!(A &amp;&amp; B)</code></td><td><code>!A || !B</code></td></tr>
<tr><td><code>!(A || B)</code></td><td><code>!A &amp;&amp; !B</code></td></tr>
</tbody>
</table>

<br>

<p>مثال: نمره‌ی معتبر یعنی <code>score &gt;= 0 &amp;&amp; score &lt;= 20</code>. پس نامعتبر یعنی <code>score &lt; 0 || score &gt; 20</code>. به عمل‌گر دقت کنید: <code>&amp;&amp;</code> تبدیل به <code>||</code> شد و هر مقایسه برعکس شد.</p>

<h4>۲) شرط نگهبان (Guard) با ارزیابی کوتاه‌مدار</h4>

<p>در بخش دوم دیدید که در <code>A &amp;&amp; B</code> اگر <code>A</code> نادرست باشد، <code>B</code> اصلاً ارزیابی نمی‌شود. حالا نتیجه‌ی الگوریتمی آن: <strong>شرطِ نگهبان را سمت چپ بگذارید تا طرف راست فقط وقتی اجرا شود که امن است.</strong> مثلاً <code>count != 0 &amp;&amp; sum / count &gt; 10</code> هیچ‌وقت تقسیم بر صفر نمی‌کند، اما اگر ترتیب را برعکس کنید، می‌کند. همین ایده را در الگوی «میانگین‌گیری» فصل بعد می‌بینید.</p>

<h4>تمرین درون فصل ۶.۲</h4>

<p><strong>نوع تمرین: Predict the Output</strong></p>

<div dir="ltr" style="margin-bottom:30px">

```c
int x = 5;
printf("%d %d\n", x > 3 && x < 10, x < 3 || x > 10);
```
</div>
<br>

<p>خروجی را پیش‌بینی کنید. سپس با دمورگان، <code>!(x &gt; 3 &amp;&amp; x &lt; 10)</code> را بدون <code>!</code> بنویسید.</p>

<h4>Example B - Real: سال کبیسه (Leap Year)</h4>
</div>

<p>قاعده‌ی تقویم میلادی چندلایه است: سالی کبیسه است که بر ۴ بخش‌پذیر باشد، <strong>مگر</strong> این‌که بر ۱۰۰ بخش‌پذیر باشد، <strong>مگر</strong> این‌که بر ۴۰۰ هم بخش‌پذیر باشد. در بخش دوم این قاعده را با یک عبارت منطقی (<code>% &amp;&amp; || !</code>) نوشتید؛ اینجا همان را به‌صورت یک <strong>زنجیره‌ی تصمیم</strong> می‌نویسیم که از <strong>خاص‌ترین استثنا</strong> به عمومی‌ترین قاعده می‌رود (۴۰۰ ← ۱۰۰ ← ۴).</p>

<div dir="ltr">

<h4>Pseudocode</h4>
</div>

<div dir="ltr" style="margin-bottom:30px">

```text
if year divisible by 400:       LEAP
else if year divisible by 100:  NOT leap
else if year divisible by 4:    LEAP
else:                           NOT leap
```
</div>
<br>

</div>

````python
%%writefile src/p03_c06_05_leap_year.c
#include <stdio.h>

int main(void)
{
    int year;

    if (scanf("%d", &year) != 1) {
        return 1;
    }

    int leap;
    if (year % 400 == 0) {
        leap = 1;
    } else if (year % 100 == 0) {
        leap = 0;
    } else {
        leap = (year % 4 == 0);
    }

    if (leap) {
        printf("%d: leap year\n", year);
    } else {
        printf("%d: common year\n", year);
    }

    return 0;
}
````

	هر سال یک <strong>مسیر متفاوت</strong> را آزمایش می‌کند: ۲۰۰۰ و ۱۶۰۰ مسیر «بخش‌پذیر بر ۴۰۰»؛ ۱۹۰۰ و ۲۱۰۰ مسیر «بر ۱۰۰ ولی نه ۴۰۰»؛ ۲۰۲۴ مسیر «بر ۴ ولی نه ۱۰۰»؛ و ۲۰۲۳ مسیر «هیچ‌کدام». به این ایده <strong>پوشش مسیرها (Path Coverage)</strong> می‌گویند: برای هر مسیر حداقل یک ورودی آزمایشی. اگر فقط ۲۰۲۴ و ۲۰۲۳ را تست می‌کردیم، پیاده‌سازی غلط <code>year % 4 == 0</code> هم قبول می‌شد.</p>

````python
%%writefile src/p03_c06_06_remainder_negative.c
#include <stdio.h>

int main(void)
{
    int n = -7;

    printf("n %% 2 = %d\n", n % 2);

    if (n % 2 == 1) {
        printf("version 1 (n %% 2 == 1): odd\n");
    } else {
        printf("version 1 (n %% 2 == 1): even   <-- wrong!\n");
    }

    if (n % 2 != 0) {
        printf("version 2 (n %% 2 != 0): odd\n");
    } else {
        printf("version 2 (n %% 2 != 0): even\n");
    }

    return 0;
}
````

<h3>6.5 زنجیره‌ی else if (انتخاب چندمسیره)</h3>

<h4>ایده الگوریتم</h4>

<p>گاهی بیش از دو مسیر داریم؛ مثلاً نمره باید به «عالی» (۱۷ تا ۲۰)، «خوب» (۱۴ تا ۱۶)، «قبول» (۱۰ تا ۱۳) یا «مردود» (زیر ۱۰) تبدیل شود. راه‌حل، <strong>زنجیره‌ای از شرط‌هاست که از بالا به پایین بررسی می‌شوند و اولین شرطی که درست باشد برنده است</strong>؛ بقیه‌ی زنجیره اصلاً بررسی نمی‌شود. این جمله‌ی کوتاه مهم‌ترین نکته‌ی این بخش است.</p>

<h4>Pseudocode</h4>
</div>

<div dir="ltr" style="margin-bottom:30px">

```text
if condition 1:      statements 1
else if condition 2: statements 2
else if condition 3: statements 3
else:                statements N     // optional: everything not matched above
```

````python
%%writefile src/p03_c06_07_grade_ladder.c
#include <stdio.h>

int main(void)
{
    int score;

    if (scanf("%d", &score) != 1) {
        return 1;
    }

    if (score < 0 || score > 20) {
        printf("%d: invalid score\n", score);
    } else if (score >= 17) {
        printf("%d: Excellent\n", score);
    } else if (score >= 14) {
        printf("%d: Good\n", score);
    } else if (score >= 10) {
        printf("%d: Pass\n", score);
    } else {
        printf("%d: Fail\n", score);
    }

    return 0;
}
````

<h4>Example B - Real: دسته‌بندی مثلث</h4>
</div>

<p>سه عدد صحیح، طول ضلع‌ها هستند. باید بگوییم معتبرند؟ مثلث می‌سازند؟ از چه نوعی (متساوی‌الاضلاع، متساوی‌الساقین، مختلف‌الاضلاع) و قائم‌الزاویه هستند یا نه؟ ایده: سؤال‌ها را به <strong>ترتیب وابستگی</strong> مرتب کنیم: تا ضلع‌ها مثبت نباشند، نامساوی مثلث بی‌معناست؛ تا نامساوی برقرار نباشد، نوع مثلث بی‌معناست.</p>

<div dir="ltr">

<h4>Pseudocode</h4>
</div>

<div dir="ltr" style="margin-bottom:30px">

```text
if a <= 0 or b <= 0 or c <= 0:      invalid
else if a+b <= c or a+c <= b or b+c <= a:   not a triangle
else:
    type ← equilateral | isosceles | scalene
    right ← a²+b² = c² or a²+c² = b² or b²+c² = a²
```
</div>
<br>

</div>

````python
%%writefile src/p03_c06_08_triangle.c
#include <stdio.h>

int main(void)
{
    int a, b, c;

    if (scanf("%d %d %d", &a, &b, &c) != 3) {
        return 1;
    }

    if (a <= 0 || b <= 0 || c <= 0) {
        printf("%d %d %d: invalid (sides must be positive)\n", a, b, c);
    } else if (a + b <= c || a + c <= b || b + c <= a) {
        printf("%d %d %d: not a triangle\n", a, b, c);
    } else {
        int right = (a * a + b * b == c * c) ||
                    (a * a + c * c == b * b) ||
                    (b * b + c * c == a * a);

        printf("%d %d %d: ", a, b, c);
        if (a == b && b == c) {
            printf("equilateral");
        } else if (a == b || b == c || a == c) {
            printf("isosceles");
        } else {
            printf("scalene");
        }
        if (right) {
            printf(", right-angled");
        }
        printf("\n");
    }

    return 0;
}
````

	در مثلث واقعی مجموع هر دو ضلع <strong>اکیداً</strong> بزرگ‌تر از ضلع سوم است. ورودی <code>1 2 3</code> (که <code>1 + 2 == 3</code>) سه نقطه روی یک خط است و یک <strong>تست مرزی</strong> عالی.</p>

<p>چرا ساختار برنامه ترکیبی از زنجیره و <code>if</code> تو در تو است؟<br>
	بخش اول (اعتبار ← مثلث بودن) یک انتخاب چندمسیره است. اما وقتی مثلث معتبر شد، <strong>دو سؤال مستقل</strong> داریم: نوع از نظر ضلع‌ها و قائم‌الزاویه بودن؛ و در یک زنجیره فقط یک مسیر اجرا می‌شود. پس پاسخ سؤال دوم را در متغیر <code>right</code> نگه می‌داریم و در انتها اضافه می‌کنیم.</p>

<h3>6.6 شرط‌های تو در تو و Guard Clause</h3>

<h4>ایده الگوریتم</h4>

<p>یک <code>if</code> می‌تواند داخل بدنه‌ی <code>if</code> یا <code>else</code> دیگری باشد؛ طبیعی است وقتی یک تصمیم <strong>فقط وقتی معنا دارد</strong> که تصمیم قبلی به نتیجه‌ی مشخصی رسیده باشد. مثال: برداشت از خودپرداز با این قواعد: مبلغ مثبت باشد؛ مضرب ۱۰ باشد؛ از موجودی (۱۰۰۰) بیشتر نباشد؛ از سقف روزانه (۵۰۰) بیشتر نباشد. نسخه‌ی تو در تو چنین است:</p>

<div dir="ltr" style="margin-bottom:30px">

```c
if (amount > 0) {
    if (amount % 10 == 0) {
        if (amount <= balance) {
            if (amount <= limit) {
                balance -= amount;                      /* OK */
            } else { /* above the daily limit */ }
        } else { /* insufficient balance */ }
    } else { /* not a multiple of 10 */ }
} else { /* amount must be positive */ }
```
</div>
<br>

<p>این کد درست است، اما دو مشکل دارد: پیام هر خطا از شرطش <strong>دور افتاده</strong> و کد به شکل <strong>هرم</strong> به راست رشد می‌کند؛ با هر سطر، خواننده باید یک شرط دیگر را به‌خاطر بسپارد. راه‌حل <strong>Guard Clause</strong> است: به‌جای «اگر معتبر بود ادامه بده»، می‌نویسیم «اگر <strong>نامعتبر</strong> است همین‌جا رد کن»؛ هر شرط، خطای خودش را کنار پیامش مدیریت می‌کند و حالت <strong>موفق</strong> در انتها می‌ماند:</p>

</div>

````python
%%writefile src/p03_c06_10_atm_guard.c
#include <stdio.h>

int main(void)
{
    int balance = 1000, limit = 500, amount;

    if (scanf("%d", &amount) != 1) {
        return 1;
    }

    if (amount <= 0) {
        printf("Denied: amount must be positive\n");
    } else if (amount % 10 != 0) {
        printf("Denied: amount must be a multiple of 10\n");
    } else if (amount > balance) {
        printf("Denied: insufficient balance\n");
    } else if (amount > limit) {
        printf("Denied: above the daily limit\n");
    } else {
        balance -= amount;
        printf("OK: withdrew %d, balance = %d\n", amount, balance);
    }

    return 0;
}
````

<h4>Example C - Tricky: Dangling Else</h4>
</div>

<p>وقتی در <code>if</code> تو در تو آکولاد نمی‌گذاریم، قاعده‌ی C این است: **هر <code>else</code> به نزدیک‌ترین <code>if</code> بدون <code>else</code> متصل می‌شود**، صرف‌نظر از تورفتگی.</p>

</div>

````python
%%writefile src/p03_c06_11_dangling_else.c
#include <stdio.h>

int main(void)
{
    int a = 1, b = 0;

    /* looks like "else belongs to the outer if", but it does not */
    if (a == 1)
        if (b == 1)
            printf("both are 1\n");
    else
        printf("else: runs when a == 1 and b != 1\n");

    /* braces make the intention explicit */
    if (a == 1) {
        if (b == 1) {
            printf("both are 1\n");
        }
    } else {
        printf("a is not 1\n");
    }

    return 0;
}
````

<h3>6.7 عمل‌گر شرطی سه‌تایی ?:</h3>

<h4>ایده الگوریتم</h4>

<p>گاهی تصمیم ما فقط برای <strong>انتخاب یک مقدار</strong> است، نه اجرای چند دستور (مثل «بزرگ‌ترین دو عدد»)؛ نوشتن <code>if-else</code> کامل برای آن پرحرف است. عمل‌گر <code>?:</code> (که در بخش دوم فقط نامش آمد) این کار را در یک عبارت انجام می‌دهد:</p>

<div dir="ltr" style="margin-bottom:30px">

```text
شرط ? مقدار_اگر_درست : مقدار_اگر_نادرست
```
</div>
<br>

<p>تفاوت بنیادین با <code>if</code>: **<code>if</code> یک دستور (Statement) است و مقدار ندارد؛ <code>?:</code> یک عبارت (Expression) است و خودش مقدار دارد.** برای همین می‌توان آن را سمت راست انتساب یا داخل <code>printf</code> گذاشت.</p>

<div dir="ltr">

<h4>Pseudocode</h4>
</div>

<div dir="ltr" style="margin-bottom:30px">

```text
result ← (condition) ? value_if_true : value_if_false
```
</div>
<br>

</div>

````python
%%writefile src/p03_c06_12_ternary.c
#include <stdio.h>

int main(void)
{
    int a = 12, b = 20;

    int max  = (a > b) ? a : b;
    int diff = (a > b) ? (a - b) : (b - a);                 /* |a - b| */
    int sign = (a - b > 0) ? 1 : ((a - b < 0) ? -1 : 0);

    printf("max = %d, |a - b| = %d, sign = %d\n", max, diff, sign);
    printf("a is %s\n", (a % 2 == 0) ? "even" : "odd");

    for (int n = 0; n <= 2; n++) {      /* a for loop: see chapter 7 */
        printf("%d file%s\n", n, (n == 1) ? "" : "s");
    }

    return 0;
}
````

````python
!gcc -std=c11 -Wall -Wextra -o bin/p03_c06_12_ternary src/p03_c06_12_ternary.c && ./bin/p03_c06_12_ternary
````

<div dir="rtl">

<p>چرا <code>?:</code> را داخل <code>printf</code> گذاشته‌ایم؟<br>
	چون <code>?:</code> مقدار تولید می‌کند (اینجا رشته‌ی <code>"even"</code> یا <code>"odd"</code>، یا پسوند جمع) و <code>printf</code> به همان مقدار نیاز دارد؛ با <code>if</code> باید دو <code>printf</code> یا یک متغیر کمکی می‌داشتیم. (حلقه‌ی <code>for</code> را در فصل بعد کامل می‌بینیم؛ فعلاً فقط بدانید <code>n</code> را از ۰ تا ۲ می‌شمارد.)</p>

<p>محاسبه‌ی <code>sign</code> با <code>?:</code> تو در تو، کار درستی است؟<br>
	از نظر عملکرد درست است، اما خواندنش از معادل <code>if-else if-else</code> سخت‌تر است؛ تو در تو کردن <code>?:</code> یعنی به مرز <strong>کاهش خوانایی</strong> رسیده‌ایم.</p>

<p>**اولویت پایین <code>?:</code>:** این عمل‌گر از تقریباً همه‌ی عمل‌گرها اولویت پایین‌تر دارد (فقط از انتساب بالاتر است)، پس بدون پرانتز نتیجه‌ی غیرمنتظره می‌دهد:</p>

<div dir="ltr" style="margin-bottom:30px">

```c
int a = 1, b = 3;
int r = a > b ? 100 : 200 + 1;      // r = 201: «+ 1» جزو شاخه‌ی نادرست است
```
</div>
<br>

<p>قاعده: هر عبارت <code>?:</code> را کامل داخل پرانتز بنویسید. اگر دو شاخه از نوع‌های متفاوت باشند (مثلاً <code>int</code> و <code>double</code>)، نتیجه به نوع بزرگ‌تر تبدیل می‌شود.</p>

<h3>6.8 switch-case</h3>
</div>

<h4>ایده الگوریتم</h4>

<p>گاهی <strong>یک مقدار را با چند مقدار ثابت</strong> مقایسه می‌کنیم: نوع عملگر ماشین‌حساب، شماره‌ی ماه، انتخاب منو. با <code>else if</code> باید عبارت <code>op == ...</code> را بارها تکرار کنیم. <code>switch</code> مقدار عبارت را <strong>یک‌بار</strong> حساب می‌کند و مستقیماً به <code>case</code> مطابق «می‌پرد»؛ اگر هیچ‌کدام نبود، به <code>default</code> می‌رود.</p>

<div dir="ltr">

<h4>Pseudocode</h4>
</div>

<div dir="ltr" style="margin-bottom:30px">

```text
evaluate expression once
jump to the case whose constant equals the value
    execute from there downwards, until "break" or the end of the switch
if no case matches: jump to default (if any)
```
</div>
<br>

<div dir="ltr">

<h4>Flowchart</h4>
</div>

<div dir="ltr" style="margin-bottom:30px">

```mermaid
flowchart TD
    A([شروع]) --> B["مقدار عبارت را یک‌بار حساب کن"]
    B --> C{"برابر case 1 ؟"}
    C -- بله --> C1["بلوک case 1"]
    C -- خیر --> D{"برابر case 2 ؟"}
    D -- بله --> D1["بلوک case 2"]
    D -- خیر --> E["بلوک default"]
    C1 -- break --> F["بعد از switch"]
    D1 -- break --> F
    E --> F
    F --> G([پایان])
```
</div>

<p><strong>قوانین:</strong> عبارت <code>switch</code> باید صحیح باشد (<code>int</code>، <code>char</code>، <code>enum</code>)، نه اعشاری یا رشته. هر <code>case</code> باید یک <strong>ثابت صحیح</strong> و منحصربه‌فرد باشد (متغیر یا بازه مثل <code>case 1..5</code> مجاز نیست). <code>switch</code> فقط <strong>تساوی</strong> را می‌سنجد. <code>default</code> اختیاری است ولی برای ورودی‌های پیش‌بینی‌نشده مفید است.</p>

<h4>Example A - Simple: ماشین‌حساب</h4>
</div>

</div>

````python
%%writefile src/p03_c06_13_switch_calculator.c
#include <stdio.h>

int main(void)
{
    int a, b;
    char op;

    if (scanf("%d %c %d", &a, &op, &b) != 3) {
        printf("Bad input\n");
        return 1;
    }

    switch (op) {
        case '+':
            printf("%d + %d = %d\n", a, b, a + b);
            break;
        case '-':
            printf("%d - %d = %d\n", a, b, a - b);
            break;
        case '*':
            printf("%d * %d = %d\n", a, b, a * b);
            break;
        case '/':
            if (b == 0) {
                printf("Error: division by zero\n");
            } else {
                printf("%d / %d = %d (remainder %d)\n", a, b, a / b, a % b);
            }
            break;
        default:
            printf("Error: unknown operator '%c'\n", op);
    }

    return 0;
}
````

<p>چرا <code>8 ^ 2</code> به <code>default</code> می‌رود؟<br>

<h4>نقش break و پدیده‌ی Fall-Through</h4>

<p>پرش به <code>case</code> مناسب فقط <strong>نقطه‌ی شروع</strong> اجراست، نه پایان آن. بعد از رسیدن به <code>case</code>، برنامه به اجرا ادامه می‌دهد و از <code>case</code> بعدی هم رد می‌شود، مگر به <code>break</code> برسد. به این رفتار <strong>Fall-Through</strong> (سقوط به بلوک بعدی) می‌گوییم.</p>

</div>

````python
%%writefile src/p03_c06_14_switch_fallthrough.c
#include <stdio.h>

int main(void)
{
    int level = 2;

    printf("--- without break ---\n");
    switch (level) {
        case 3:
            printf("A ");
        case 2:
            printf("B ");
        case 1:
            printf("C ");
        default:
            printf("D ");
    }
    printf("\n");

    printf("--- with break ---\n");
    switch (level) {
        case 3:
            printf("A ");
            break;
        case 2:
            printf("B ");
            break;
        case 1:
            printf("C ");
            break;
        default:
            printf("D ");
    }
    printf("\n");

    return 0;
}
````

<h4>Example B - Real: تعداد روزهای ماه</h4>
</div>

<p>ترکیب دو الگوریتم قبلی: ماه‌ها سه گروه‌اند (۳۰ روزه، ۳۱ روزه و فوریه) و فوریه خودش یک تصمیم است: همان زنجیره‌ی سال کبیسه‌ی 6.4.</p>

</div>

````python
%%writefile src/p03_c06_15_days_in_month.c
#include <stdio.h>

int main(void)
{
    int month, year, days = 0;

    if (scanf("%d %d", &month, &year) != 2) {
        return 1;
    }

    switch (month) {
        case 4: case 6: case 9: case 11:
            days = 30;
            break;
        case 2:
            if (year % 400 == 0) {
                days = 29;
            } else if (year % 100 == 0) {
                days = 28;
            } else {
                days = (year % 4 == 0) ? 29 : 28;
            }
            break;
        case 1: case 3: case 5: case 7: case 8: case 10: case 12:
            days = 31;
            break;
        default:
            printf("Invalid month: %d\n", month);
            return 1;
    }

    printf("Month %d of year %d has %d days\n", month, year, days);
    return 0;
}
````

````python
%%writefile src/p03_c06_16_switch_score_div10.c
#include <stdio.h>

int main(void)
{
    int score;
    char grade = '?';

    if (scanf("%d", &score) != 1) {
        return 1;
    }

    if (score < 0 || score > 100) {
        printf("%d: invalid score\n", score);
        return 1;
    }

    switch (score / 10) {
        case 10:            /* only 100 */
        case 9:
            grade = 'A';
            break;
        case 8:
            grade = 'B';
            break;
        case 7:
            grade = 'C';
            break;
        case 6:
            grade = 'D';
            break;
        default:
            grade = 'F';
    }

    printf("%d -> %c\n", score, grade);
    return 0;
}
````

<p>سطر آخر پنهان‌ترین تفاوت است. فرض کنید عضو باشگاه ۵٪ و سفارش بالای ۱۰۰۰ مقدار ۱۰٪ تخفیف می‌گیرد، و <strong>هر دو قابل جمع‌اند</strong>. اگر این دو شرط را به هم بچسبانیم، فقط یکی اعمال می‌شود:</p>

````python
%%writefile src/p03_c06_17_independent_vs_chain.c
#include <stdio.h>

int main(void)
{
    int total, member;

    if (scanf("%d %d", &total, &member) != 2) {
        return 1;
    }

    int independent = 0;               /* both discounts may apply */
    if (member) {
        independent += 5;
    }
    if (total >= 1000) {
        independent += 10;
    }

    int chained = 0;                   /* only the first match applies */
    if (member) {
        chained = 5;
    } else if (total >= 1000) {
        chained = 10;
    }

    printf("total=%d member=%d | independent: %d%% | chained: %d%%\n",
           total, member, independent, chained);
    return 0;
}
````

<h3>6.10 خطاهای رایج در ساختارهای شرطی</h3>

<p><code>=</code> به‌جای <code>==</code> و مقایسه‌ی دقیق اعداد اعشاری را در بخش‌های اول و دوم دیدید و اینجا تکرار نمی‌کنیم. اشتباه‌های دیگری که خاص شرط‌هاست:</p>

<p><strong>❌ اشتباه ۱: نقطه‌ویرگول بعد از شرط</strong></p>

<div dir="ltr" style="margin-bottom:30px">

```c
if (score >= 18);
{
    printf("Excellent\n");
}
```
</div>
<br>

<p>چرا اشتباه است؟<br>
	<code>;</code> یک «دستور خالی» است و بدنه‌ی <code>if</code> همان می‌شود؛ بلوک بعدی مستقل است و همیشه اجرا می‌شود.</p>

<p><strong>✅ نسخه صحیح:</strong> <code>;</code> را حذف کنید.</p>

<p><strong>❌ اشتباه ۲: مقایسه‌ی زنجیره‌ای</strong></p>

<div dir="ltr" style="margin-bottom:30px">

```c
if (0 <= score <= 20) { ... }       // همیشه درست است
```
</div>
<br>

<p>چرا اشتباه است؟<br>
	<code>0 &lt;= score</code> نتیجه‌ی <code>0</code> یا <code>1</code> است و بعد آن عدد با <code>20</code> مقایسه می‌شود. <strong>✅ نسخه صحیح:</strong> <code>score &gt;= 0 &amp;&amp; score &lt;= 20</code>.</p>

<p>**❌ اشتباه ۳: <code>x == 1 || 2</code>**</p>

<div dir="ltr" style="margin-bottom:30px">

```c
if (x == 1 || 2) { ... }            // همیشه درست است
```
</div>
<br>

<p>چرا اشتباه است؟<br>
	عبارت <code>(x == 1) || 2</code> خوانده می‌شود و <code>2</code> غیرصفر است. <strong>✅ نسخه صحیح:</strong> <code>x == 1 || x == 2</code>.</p>

<h4>Example B - Real: جمع ارقام یک عدد</h4>

<p>مجموع ارقام <code>n</code> را می‌خواهیم (<code>472</code> ← <code>4+7+2</code>). رقم آخر را با <code>n % 10</code> جدا می‌کنیم و با <code>n / 10</code> حذف می‌کنیم؛ پس تا وقتی عدد صفر نشده، رقم آخر را بردار و به مجموع اضافه کن. <strong>تعداد تکرار از قبل معلوم نیست</strong> (به تعداد ارقام بستگی دارد)، پس <code>while</code> طبیعی‌ترین انتخاب است.</p>

<div dir="ltr">

<h4>Pseudocode</h4>

<div dir="ltr" style="margin-bottom:30px">

```text
sum ← 0, count ← 0
while n > 0:
    sum ← sum + (n mod 10)
    n ← n div 10
    count ← count + 1
```

````python
%%writefile src/p03_c07_03_digit_sum.c
#include <stdio.h>

int main(void)
{
    int n;

    if (scanf("%d", &n) != 1 || n < 0) {
        printf("Please enter a non-negative integer\n");
        return 1;
    }

    int original = n;
    int sum = 0, count = 0;

    while (n > 0) {
        sum += n % 10;
        n /= 10;
        count++;
    }

    printf("%d: digit sum = %d, digit count = %d\n", original, sum, count);
    return 0;
}
````

<h4>Example C - Tricky: الگوی Sentinel</h4>
</div>

<p>گاهی تعداد داده از قبل معلوم نیست و کاربر پایان داده‌ها را با یک مقدار ویژه اعلام می‌کند؛ این مقدار را <strong>Sentinel</strong> (نگهبان) می‌نامیم (اینجا <code>-1</code> که نمره‌ی معتبر نیست). <code>scanf(...) == 1</code> هم تضمین می‌کند ورودی واقعاً خوانده شد و در پایان ورودی حلقه خودبه‌خود تمام می‌شود.</p>

</div>

````python
%%writefile src/p03_c07_04_while_sentinel_avg.c
#include <stdio.h>

int main(void)
{
    int x, count = 0, sum = 0;

    while (scanf("%d", &x) == 1 && x != -1) {
        sum += x;
        count++;
    }

    if (count == 0) {
        printf("No scores entered\n");
    } else {
        printf("count = %d, sum = %d, average = %.2f\n",
               count, sum, (double)sum / count);
    }

    return 0;
}
````

<p>ورودی سوم (<code>14</code> بدون <code>-1</code>) چرا درست کار می‌کند؟<br>
	با تمام شدن ورودی <code>scanf</code> مقدار <code>1</code> برنمی‌گرداند و شرط حلقه نادرست می‌شود؛ و چون <code>&amp;&amp;</code> کوتاه‌مدار است، <code>x</code> فقط وقتی با <code>-1</code> مقایسه می‌شود که واقعاً مقدار گرفته باشد (همان شرط نگهبان 6.2).</p>

<h3>7.3 حلقه‌ی do-while</h3>

<h4>ایده الگوریتم</h4>

<p>گاهی بدنه باید <strong>حداقل یک‌بار</strong> اجرا شود و بعد بپرسیم ادامه بدهیم یا نه. مثال کلاسیک: گرفتن ورودی معتبر؛ پیش از این‌که اعتبارش را بسنجیم، باید مقداری گرفته باشیم. <code>do-while</code> شرط را <strong>بعد از هر تکرار</strong> بررسی می‌کند (<strong>Post-test Loop</strong>).</p>

<div dir="ltr">

<h4>Pseudocode</h4>

```text

```

````python
%%writefile src/p03_c07_06_do_while_validation.c
#include <stdio.h>

int main(void)
{
    int s = -1;

    do {
        printf("Enter score (0-20): ");
        int got = scanf("%d", &s);

        if (got == EOF) {
            printf("\nNo more input\n");
            return 1;
        }
        if (got == 0) {                  /* text instead of a number */
            printf("Not a number\n");
            scanf("%*s");                /* throw the bad word away */
            s = -1;
        } else if (s < 0 || s > 20) {
            printf("Invalid: %d\n", s);
        }
    } while (s < 0 || s > 20);

    printf("Accepted: %d\n", s);
    return 0;
}
````

````python
%%writefile src/p03_c07_07_for_forms.c
#include <stdio.h>

int main(void)
{
    printf("Up:     ");
    for (int i = 1; i <= 5; i++) {
        printf("%d ", i);
    }

    printf("\nDown:   ");
    for (int i = 5; i >= 1; i--) {
        printf("%d ", i);
    }

    printf("\nStep 2: ");
    for (int i = 0; i <= 10; i += 2) {
        printf("%d ", i);
    }

    printf("\nTwo variables: ");
    for (int lo = 0, hi = 6; lo < hi; lo++, hi--) {
        printf("(%d,%d) ", lo, hi);
    }
    printf("\n");

    return 0;
}
````

<h4>تعداد تکرارها: خطای یکی‌کم‌یکی‌زیاد (Off-by-One)</h4>

````python
%%writefile src/p03_c07_08_float_counter.c
#include <stdio.h>

int main(void)
{
    /* Danger: counting with a double and testing with != */
    double x = 0.0;
    int k = 0;
    while (x != 1.0 && k < 12) {     /* k < 12 is a safety cap for this demo */
        x += 0.1;
        k++;
    }
    printf("after %d additions: x = %.17f, x == 1.0 ? %d\n", k, x, x == 1.0);

    /* Safer: count with an integer, derive the double from it */
    int hits = 0;
    for (int j = 0; j <= 10; j++) {
        double y = j * 0.1;
        if (y == 1.0) {
            hits++;
        }
    }
    printf("integer counter: reached 1.0 exactly %d time(s)\n", hits);

    return 0;
}
````

````python
%%writefile src/p03_c07_09_counter_accumulator.c
#include <stdio.h>

int main(void)
{
    int count = 0;
    int sum = 0;

    for (int i = 1; i < 1000; i++) {
        if (i % 3 == 0 || i % 5 == 0) {
            count++;
            sum += i;
        }
    }

    printf("count = %d, sum = %d\n", count, sum);
    return 0;
}
````

<h4>Min / Max</h4>
</div>

<p>اگر همه‌ی داده‌ها را هم‌زمان در حافظه نداریم، حین خواندن «بهترین تا اینجا» را نگه می‌داریم. نکته‌ی کلیدی: مقدار اولیه را <strong>با اولین داده‌ی واقعی</strong> بسازید، نه با عددی دلخواه مثل <code>0</code>؛ وگرنه اگر همه‌ی داده‌ها منفی باشند جواب غلط می‌شود.</p>

</div>

````python
%%writefile src/p03_c07_10_min_max.c
#include <stdio.h>

int main(void)
{
    int x, min = 0, max = 0, count = 0;

    while (scanf("%d", &x) == 1) {
        if (count == 0) {
            min = x;
            max = x;
        } else {
            if (x < min) {
                min = x;
            }
            if (x > max) {
                max = x;
            }
        }
        count++;
    }

    if (count == 0) {
        printf("No data\n");
    } else {
        printf("count = %d, min = %d, max = %d\n", count, min, max);
    }

    return 0;
}
````

<h4>Search + Flag: آیا n اول است؟</h4>
</div>

<p>عدد اول، عددی بزرگ‌تر از ۱ است که فقط بر ۱ و خودش بخش‌پذیر است. برای <code>d</code> از ۲ به بعد ببین آیا <code>n</code> بر <code>d</code> بخش‌پذیر است؛ <strong>اگر حتی یک مقسوم‌علیه پیدا شد، جواب «اول نیست» است و ادامه‌دادن بی‌معنی است.</strong> پس متغیر <code>is_prime</code> را با فرض «اول است» می‌سازیم و در صورت یافتن شاهد مخالف، آن را عوض و حلقه را با <code>break</code> قطع می‌کنیم (<strong>الگوی پرچم</strong>). بهبود مهم: اگر <code>n = a × b</code> باشد، دست‌کم یکی از <code>a</code> و <code>b</code> از <code>√n</code> بزرگ‌تر نیست؛ پس کافی است <code>d</code> را فقط تا <code>√n</code> بررسی کنیم (شرط <code>d * d &lt;= n</code>).</p>

<div dir="ltr">

<h4>Pseudocode</h4>
</div>

<div dir="ltr" style="margin-bottom:30px">

```text
if n < 2: NOT PRIME
is_prime ← TRUE
for d from 2 while d × d ≤ n:
    if n mod d = 0:
        is_prime ← FALSE
        break
```
</div>
<br>

</div>

````python
%%writefile src/p03_c07_11_prime_check.c
#include <stdio.h>

int main(void)
{
    int n;

    if (scanf("%d", &n) != 1) {
        return 1;
    }

    if (n < 2) {
        printf("%d: not prime (must be at least 2)\n", n);
        return 0;
    }

    /* Version 1: try every d from 2 to n-1 */
    int prime_slow = 1, steps_slow = 0;
    for (int d = 2; d < n; d++) {
        steps_slow++;
        if (n % d == 0) {
            prime_slow = 0;
            break;
        }
    }

    /* Version 2: try d only while d*d <= n */
    int prime_fast = 1, steps_fast = 0;
    for (int d = 2; (long long)d * d <= n; d++) {
        steps_fast++;
        if (n % d == 0) {
            prime_fast = 0;
            break;
        }
    }

    printf("%d: %s | steps: slow = %d, fast = %d%s\n", n,
           prime_fast ? "prime" : "not prime", steps_slow, steps_fast,
           prime_slow != prime_fast ? "  MISMATCH!" : "");
    return 0;
}
````

<h4>Convergence: الگوریتم اقلیدس (ب.م.م)</h4>
</div>

<p>ایده: <code>gcd(a, b) = gcd(b, a mod b)</code>؛ مسئله را هر بار به مسئله‌ی <strong>کوچک‌تری</strong> با همان جواب تبدیل می‌کنیم تا به <code>b = 0</code> برسیم که جواب همان <code>a</code> است. شرط ادامه <code>b != 0</code> است و هر تکرار <code>b</code> را اکیداً کوچک‌تر می‌کند (چون <code>a mod b &lt; b</code>).</p>

<div dir="ltr">

<h4>Pseudocode</h4>
</div>

<div dir="ltr" style="margin-bottom:30px">

```text
while b ≠ 0:
    r ← a mod b
    a ← b
    b ← r
return a
```
</div>
<br>

</div>

````python
%%writefile src/p03_c07_12_euclid_gcd.c
#include <stdio.h>

int main(void)
{
    int a, b;

    if (scanf("%d %d", &a, &b) != 2 || a < 0 || b < 0) {
        printf("Please enter two non-negative integers\n");
        return 1;
    }

    int x = a, y = b, steps = 0;
    while (y != 0) {
        int r = x % y;
        x = y;
        y = r;
        steps++;
    }

    printf("gcd(%d, %d) = %d  (%d step(s))\n", a, b, x, steps);
    return 0;
}
````

<p>از کجا بدانیم حلقه حتماً تمام می‌شود؟<br>
	باید یک <strong>کمیت نزولی</strong> پیدا کنیم که در هر تکرار اکیداً کم شود و هرگز از صفر پایین‌تر نرود؛ این اثبات پایان‌پذیری حلقه است. اینجا <code>b</code> چنین کمیتی است. حلقه‌ای که چنین کمیتی نداشته باشد، «شاید» تمام نشود.</p>

<p>الگوی دیگری از Convergence جایی است که جواب دقیق نداریم و فقط می‌توانیم «هر بار بهتر» شویم. برای <code>√x</code> از حدس <code>g</code> شروع می‌کنیم و آن را با <code>g = (g + x / g) / 2</code> بهبود می‌دهیم؛ شرط توقف «برابر شدن» نیست (در اعداد اعشاری قابل اعتماد نیست) بلکه «خطا از یک <strong>تلورانس</strong> کوچک‌تر شد». هر حلقه‌ی همگرا یک <strong>سقف تکرار</strong> هم می‌خواهد تا اگر فرض‌های روش نقض شد، برنامه برای همیشه نایستد (قدر مطلق را فعلاً با <code>?:</code> می‌نویسیم؛ تابع آماده‌ی <code>fabs</code> هم در <code>math.h</code> هست):</p>

</div>

````python
%%writefile src/p03_c07_13_newton_sqrt.c
#include <stdio.h>

int main(void)
{
    double x = 2.0;
    double g = x;                 /* initial guess */
    double eps = 1e-9;
    int steps = 0;

    double err = g * g - x;
    err = (err < 0) ? -err : err;          /* absolute value via ?: */

    while (err > eps && steps < 50) {      /* 50 = safety cap */
        g = (g + x / g) / 2.0;
        err = g * g - x;
        err = (err < 0) ? -err : err;
        steps++;
        printf("step %d: g = %.10f, error = %.3e\n", steps, g, err);
    }

    printf("sqrt(%.1f) ~ %.10f after %d step(s)\n", x, g, steps);
    return 0;
}
````

<p>در بخش دوم رمز سزار را فقط برای یک حرف دیدید و گفتیم نسخه‌ی کامل بعد از <code>if</code> و حلقه می‌آید. حالا یک خط ورودی را کاراکتر به کاراکتر می‌خوانیم؛ هر حرف را به اندازه‌ی <code>shift</code> جلو می‌بریم و با <code>% 26</code> از <code>z</code> به <code>a</code> برمی‌گردانیم و بقیه‌ی کاراکترها را دست‌نخورده چاپ می‌کنیم. دقت کنید حرف‌های کوچک و بزرگ <strong>بازه‌اند</strong> (پس <code>if</code> لازم است نه <code>switch</code>).</p>

````python
%%writefile src/p03_c07_14_caesar.c
#include <stdio.h>

int main(void)
{
    int shift = 3;
    char ch;

    /* read one line character by character: until newline or end of input */
    while (scanf("%c", &ch) == 1 && ch != '\n') {
        if (ch >= 'a' && ch <= 'z') {
            ch = 'a' + (ch - 'a' + shift) % 26;
        } else if (ch >= 'A' && ch <= 'Z') {
            ch = 'A' + (ch - 'A' + shift) % 26;
        }
        printf("%c", ch);
    }
    printf("\n");

    return 0;
}
````

````python
%%writefile src/p03_c07_15_nested_triangle.c
#include <stdio.h>

int main(void)
{
    printf("Rectangle (3 x 5):\n");
    for (int i = 0; i < 3; i++) {
        for (int j = 0; j < 5; j++) {
            printf("*");
        }
        printf("\n");
    }

    int n = 4;
    printf("\nRight triangle (n = %d):\n", n);
    for (int i = 1; i <= n; i++) {
        for (int j = 1; j <= i; j++) {
            printf("*");
        }
        printf("\n");
    }

    return 0;
}
````

<h3>7.8 حلقه‌های بی‌نهایت (عمدی و سهوی)</h3>

<p><strong>بی‌نهایت سهوی:</strong> حلقه‌ای که شرطش هرگز نادرست نمی‌شود. چهار علت رایج: (۱) فراموش‌کردن به‌روزرسانی، (۲) به‌روزرسانی در جهت اشتباه، (۳) شرطی که با نوع داده جور نیست (شمارنده‌ی <code>unsigned</code> با <code>i &gt;= 0</code>، یا <code>!=</code> روی اعشاری)، (۴) <code>continue</code> پیش از به‌روزرسانی در <code>while</code>. راه کشف: مقدار متغیرهای شرط را داخل حلقه چاپ کنید و ببینید به سمت پایان حرکت می‌کنند یا نه (همان <code>printf</code> ردیابی بخش اول).</p>

<p><strong>بی‌نهایت عمدی:</strong> گاهی «تا ابد تکرار کن مگر وقتی که...» بهترین شکل مسئله است (منو، سرور). شرط را ثابت درست (<code>while (1)</code>) می‌گذاریم و <strong>از داخل بدنه</strong> با <code>break</code> (یا <code>return</code>) خارج می‌شویم:</p>

</div>

````python
%%writefile src/p03_c07_19_infinite_loop_break.c
#include <stdio.h>

int main(void)
{
    int x, total = 0;

    while (1) {
        if (scanf("%d", &x) != 1) {
            printf("input ended\n");
            break;
        }
        if (x == 0) {
            printf("zero received: stop\n");
            break;
        }
        total += x;
        printf("running total = %d\n", total);
    }

    printf("final total = %d\n", total);
    return 0;
}
````

````python
%%writefile src/p03_c07_20_loop_pitfalls.c
#include <stdio.h>

int main(void)
{
    /* 1. off-by-one: < versus <= */
    int sum_lt = 0, sum_le = 0, it_lt = 0, it_le = 0;
    for (int i = 1; i < 10; i++)  { sum_lt += i; it_lt++; }
    for (int i = 1; i <= 10; i++) { sum_le += i; it_le++; }
    printf("1) i < 10  : %d iterations, sum = %d\n", it_lt, sum_lt);
    printf("   i <= 10 : %d iterations, sum = %d\n", it_le, sum_le);

    /* 2. unsigned counter: i >= 0 is always true (capped here at 5 iterations) */
    printf("2) unsigned countdown from 2:");
    int guard = 0;
    for (unsigned int u = 2; u >= 0 && guard < 5; u--, guard++) {
        printf(" %u", u);
    }
    printf("\n");

    /* 3. a stray semicolon turns the body into an empty statement */
    int count = 0;
    for (int i = 0; i < 5; i++);
    {
        count++;
    }
    printf("3) count = %d (expected 5)\n", count);

    return 0;
}
````

<div dir="ltr">

<h4>goto چیست؟</h4>

<p>زبان C دستوری دارد که کنترل برنامه را <strong>بدون هیچ ساختار</strong> به یک <strong>برچسب (Label)</strong> دلخواه در همان تابع می‌برد (<code>goto name;</code> ... <code>name:</code>). هر ساختاری که تا اینجا دیدید را می‌توان با <code>goto</code> و <code>if</code> ساخت؛ کامپایلر در پایین‌ترین سطح همین کار را می‌کند. یک حلقه‌ی جمع ۱ تا ۵ با <code>goto</code> و همان حلقه با <code>while</code>:</p>

````python
%%writefile src/p03_c07_21_goto_vs_while.c
#include <stdio.h>

int main(void)
{
    /* version 1: a loop built from goto */
    int i = 1, sum1 = 0;
loop:
    if (i > 5) {
        goto done;
    }
    sum1 += i;
    i++;
    goto loop;
done:
    printf("goto  version: sum = %d\n", sum1);

    /* version 2: the same loop with while */
    int sum2 = 0;
    int k = 1;
    while (k <= 5) {
        sum2 += k;
        k++;
    }
    printf("while version: sum = %d\n", sum2);

    return 0;
}
````

<h4>جایگزین‌های goto</h4>

<p>وسوسه‌ی رایج <code>goto</code>، <strong>خروج از حلقه‌های تودرتو</strong> است. راه‌های ساخت‌یافته: یک <strong>پرچم</strong>، یا انتقال حلقه‌ها به <strong>تابع</strong> و <code>return</code> (در بخش بعد). برنامه‌ی زیر اولین جفت <code>(i, j)</code> با <code>i * j == 24</code> و <code>i + j == 10</code> را با <code>goto</code> و با پرچم پیدا می‌کند:</p>

</div>

````python
%%writefile src/p03_c07_22_nested_exit.c
#include <stdio.h>

int main(void)
{
    /* Version 1: goto out of both loops */
    for (int i = 1; i <= 9; i++) {
        for (int j = 1; j <= 9; j++) {
            if (i * j == 24 && i + j == 10) {
                printf("goto version: i = %d, j = %d\n", i, j);
                goto found;
            }
        }
    }
    printf("goto version: not found\n");
found:

    /* Version 2: a flag in both loop conditions */
    int fi = 0, fj = 0, done = 0;
    for (int i = 1; i <= 9 && !done; i++) {
        for (int j = 1; j <= 9 && !done; j++) {
            if (i * j == 24 && i + j == 10) {
                fi = i;
                fj = j;
                done = 1;
            }
        }
    }
    if (done) {
        printf("flag version: i = %d, j = %d\n", fi, fj);
    } else {
        printf("flag version: not found\n");
    }

    return 0;
}
````

<h3>7.10 هزینه‌ی حلقه‌ها و مرتبه‌ی رشد (Big-O)</h3>

<p>برای مقایسه‌ی الگوریتم‌ها معیاری لازم است که به سرعت کامپیوتر وابسته نباشد. ساده‌ترین معیار: <strong>بدنه‌ی حلقه چند بار اجرا می‌شود؟</strong> این عدد را به‌صورت تابعی از اندازه‌ی ورودی <code>n</code> می‌نویسیم. در بخش اول دیدید جست‌وجوی دودویی برای ۱ میلیون حالت فقط حدود ۲۰ حدس می‌خواهد؛ حالا زبان دقیق‌تری برای گفتنش داریم.</p>

</div>

````python
%%writefile src/p03_c07_23_iteration_counts.c
#include <stdio.h>

int main(void)
{
    int n = 1000;

    int linear = 0;
    for (int i = 0; i < n; i++) {
        linear++;
    }

    int halving = 0;
    for (int m = n; m > 1; m /= 2) {
        halving++;
    }

    long long quadratic = 0;
    for (int i = 0; i < n; i++) {
        for (int j = 0; j < n; j++) {
            quadratic++;
        }
    }

    printf("n = %d\n", n);
    printf("single loop      : %d iterations\n", linear);
    printf("halving loop     : %d iterations\n", halving);
    printf("two nested loops : %lld iterations\n", quadratic);

    return 0;
}
````

<h4>مرتبه‌ی رشد و نماد O</h4>

<p><strong>مرتبه‌ی رشد</strong> یعنی وقتی <code>n</code> بزرگ می‌شود، تعداد تکرارها چگونه بزرگ می‌شود. در <strong>نماد O (Big-O)</strong> فقط به همین رفتار نگاه می‌کنیم و ضریب‌های ثابت و جمله‌های کوچک‌تر را کنار می‌گذاریم: تعداد تکرارهای مثلث <code>n(n+1)/2</code> است و می‌گوییم <code>O(n²)</code>؛ حلقه‌ای که ۳ دستور در بدنه دارد هم همان <code>O(n)</code> است. ترتیب رایج‌ها از سریع به کند:</p>

<p><strong>بهترین، بدترین و میانگین حالت:</strong> تعداد تکرار گاهی به <strong>خود داده</strong> هم بستگی دارد، نه فقط به <code>n</code>. در آزمون اول بودن، عدد زوج در همان گام اول رد می‌شود (<strong>بهترین حالت</strong>) ولی عدد اول تا آخر می‌رود (<strong>بدترین حالت</strong>). معمولاً <strong>بدترین حالت</strong> را گزارش می‌کنیم، چون یک سقف مطمئن به ما می‌دهد؛ <strong>حالت میانگین</strong> رفتار معمول روی داده‌های «تصادفی» است. در بخش‌های بعد (جستجو و مرتب‌سازی) از همین زبان استفاده می‌کنیم.</p>

<p>هر حلقه چند بار بدنه را اجرا می‌کند (برحسب <code>n</code>) و از چه مرتبه‌ای است؟</p>

<h3>7.11 خطاهای رایج و کاربردهای واقعی</h3>

<ul>
<li><strong>Off-by-One:</strong> اشتباه <code>&lt;</code> و <code>&lt;=</code> یا شروع از ۰ و ۱. درمان: پیش از نوشتن، تعداد تکرار را با فرمول بازه بسنجید.</li>
<li><strong>فراموش‌کردن یا اشتباه‌گرفتن به‌روزرسانی:</strong> درمان: بپرسید «کدام کمیت هر بار به پایان نزدیک‌تر می‌شود؟»</li>
<li>**نقطه‌ویرگول بعد از <code>for</code> یا <code>while</code><strong>، </strong>شمارنده‌ی اعشاری یا <code>unsigned</code><strong>، </strong>تغییر شمارنده داخل بدنه‌ی <code>for</code><strong> و </strong><code>continue</code> پیش از به‌روزرسانی در <code>while</code>**.</li>
<li><strong>مقداردهی اولیه‌ی اشتباه:</strong> <code>max = 0</code> به‌جای اولین داده، یا <code>prime = 1</code> بیرون از حلقه‌ی بیرونی.</li>
<li><strong>سرریز در Accumulator</strong> و <strong>ورودی خالی</strong> (تقسیم بر <code>count</code> وقتی <code>count = 0</code>).</li>
</ul>

<p>کاربردها: پردازش همه‌ی رکوردهای یک پایگاه‌داده یا خطوط یک فایل؛ حلقه‌ی اصلی بازی‌ها و برنامه‌های گرافیکی (ورودی ← به‌روزرسانی ← رسم)؛ شبیه‌سازی‌های علمی و روش‌های عددی که تا رسیدن به دقت مطلوب تکرار می‌شوند؛ سرورها و سیستم‌های نهفته که برای همیشه منتظر رویداد می‌مانند (<code>while (1)</code> عمدی).</p>

<h3>7.12 جمع‌بندی فصل ۷</h3>

<p>در این فصل دیدیم تکرار دستی دستورها با رشد مسئله ناممکن می‌شود و <strong>حلقه</strong> با چهار بخش (مقداردهی، شرط، بدنه، به‌روزرسانی) آن را حل می‌کند. <code>while</code> (شرط اول)، <code>do-while</code> (شرط آخر، حداقل یک‌بار) و <code>for</code> (شمارنده‌ی مشخص) را شناختیم و دیدیم انتخاب بین آن‌ها گاهی درستی الگوریتم را تعیین می‌کند. با هفت <strong>الگوی حلقه</strong> دیدیم بیشتر مسئله‌ها ترکیبی از چند الگوی آشنا هستند. حلقه‌های تودرتو را برای داده‌های دوبعدی به کار بردیم و هزینه‌شان را ضرب کردیم. با <code>break</code> و <code>continue</code> رفتار حلقه را در میانه‌ی تکرار تغییر دادیم، خطر حلقه‌ی بی‌نهایت را دیدیم و دلیل پرهیز از <code>goto</code> را از دید خوانایی و استدلال‌پذیری بررسی کردیم. در پایان تکرارها را شمردیم و با نماد <code>O</code> الگوریتم‌ها را مقایسه کردیم.</p>

<p><strong>راهنمایی 1:</strong></p>

<p>برای آمار بدون آرایه، پیش از حلقه‌ی منو این متغیرها را تعریف کنید: <code>count</code>، <code>sum</code>، <code>min</code>، <code>max</code> و چهار شمارنده‌ی دسته. <code>min</code> و <code>max</code> را با <strong>اولین نمره</strong> مقداردهی کنید (از <code>count == 0</code> بفهمید اولین نمره است).</p>

<p><strong>راهنمایی 2:</strong></p>

<p>چون منو باید حداقل یک‌بار نمایش داده شود، <code>do { ... } while (choice != 4);</code> مناسب است. مراقب باشید <code>scanf</code> در پایان ورودی شکست می‌خورد و متن به‌جای عدد در ورودی می‌ماند (7.3)؛ این حالت‌ها را مدیریت کنید تا حلقه بی‌نهایت نشود.</p>

<p><strong>راهنمایی 3:</strong></p>

<p>برای هیستوگرام برای هر دسته یک <code>printf</code> نام دسته و یک <code>for</code> ساده برای ستاره‌ها بنویسید؛ کد تکراری فعلاً قابل قبول است (در بخش توابع راه تمیزتری یاد می‌گیرید و در بخش پنجم با آرایه‌ها این سناریو کامل می‌شود).</p>

<p><strong>راهنمایی 4:</strong></p>

<p>حالت‌های مرزی را تست کنید: خروج بدون هیچ نمره، آمار پیش از ثبت نمره، نمره‌های <code>0</code> و <code>20</code>، و <code>9</code> و <code>10</code>، <code>13</code> و <code>14</code>، <code>16</code> و <code>17</code>.</p>

<h3>بخش بعد</h3>

<p>حالا که می‌توانیم تصمیم بگیریم و تکرار کنیم، برنامه‌ها بلندتر می‌شوند و کپی‌کردن یک تکه کد (مثل چاپ هیستوگرام برای چهار دسته) آزارنده است. در بخش چهارم یاد می‌گیریم یک تکه کد را زیر یک نام بسته‌بندی کنیم (تابع) و تابعی بنویسیم که خودش را صدا بزند (بازگشت).</p>

<br>

<h2>نگاهی به یک دانشمند تأثیرگذار در این حوزه</h2>

<p>خب، حالا که با ساختارهای تصمیم و تکرار آشنا شدیم و دیدیم چرا از <code>goto</code> پرهیز می‌کنیم، بد نیست با دانشمندی آشنا شویم که این نگاه را در علوم کامپیوتر جا انداخت.</p>

<p><strong>ادسخر ویبه دایکسترا (Edsger W. Dijkstra)</strong> دانشمند علوم کامپیوتر هلندی (۱۹۳۰ تا ۲۰۰۲) بود. او الگوریتم معروف <strong>کوتاه‌ترین مسیر</strong> در گراف‌ها را (که ریشه‌ی بسیاری از نقشه‌یاب‌های امروزی است) در دهه‌ی ۱۹۵۰ ابداع کرد. اما شاید تأثیرگذارترین کار او در نگاه ما به «<strong>ساختار برنامه</strong>» باشد: در سال ۱۹۶۸ نامه‌ی کوتاه او درباره‌ی دستور <code>goto</code> منتشر شد و بحثی را آغاز کرد که به <strong>برنامه‌نویسی ساخت‌یافته (Structured Programming)</strong> انجامید. استدلال او فنی بود، نه سلیقه‌ای: اگر نتوانیم از روی <strong>متن برنامه</strong> رفتار آن در <strong>زمان اجرا</strong> را پیش‌بینی کنیم، نمی‌توانیم درستی‌اش را اثبات یا حتی با اطمینان بررسی کنیم. او در سال ۱۹۷۲ جایزه‌ی <strong>تورینگ</strong> را دریافت کرد.</p>

<p>مسیر فکری دایکسترا با کاری که در این بخش کردیم دور نیست: <strong>یک برنامه‌ی پیچیده ← تشخیص این‌که فهمش سخت است ← پیدا کردن ساختارهایی که استدلال درباره‌ی آن‌ها آسان است (یک ورودی، یک خروجی) ← ساختن همه‌چیز از همان‌ها</strong>. نامتغیر حلقه، شرط توقف و Trace که تمرین کردید، همه ابزارهای همین نگاه‌اند.</p>

<blockquote><p><strong>شاید یکی از نکات جالب این داستان این باشد که یکی از مؤثرترین ایده‌های برنامه‌نویسی، نه افزودن یک قابلیت جدید، بلکه «محدود کردن» قدرت برنامه‌نویس به ساختارهای قابل فهم بود.</strong></p></blockquote>
