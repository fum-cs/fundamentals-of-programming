# توابع و بازگشت
### از کلی‌ترین تعریف ممکن، تا ریزترین جزئیات نحوی — یک روایتِ از-کل-به-جزء

> **یادداشت قالب:** برخلاف نوت‌بوک قبلی (پوینتر) که از یک مثال ملموس شروع می‌شد، این‌بار روند تدریس عمداً **از انتزاعی‌ترین سطح ممکن به سمت جزئی‌ترین نحو زبان C** حرکت می‌کند. هر «سطح» جدید، سطح قبلی را *محدودتر و مشخص‌تر* می‌کند، نه اینکه چیز نامرتبطی اضافه کند. قبل از هر بلوک کد C، معنایِ (سمانتیک) مفهوم را در سطحِ حافظه/پردازنده یا ریاضیات توضیح می‌دهیم؛ نحو C همیشه آخرین چیزی است که می‌رسد.


## 🎯 اهداف یادگیری
پس از این جلسه دانشجو باید بتواند:

1. تعریف انتزاعی «تابع» را مستقل از هر زبان برنامه‌نویسی بیان کند.
2. تعریف رسمی ریاضیِ تابع (دامنه، برد، تک‌مقداری بودن) را توضیح دهد و از آن نتیجه بگیرد چرا یک تابع دقیقاً یک خروجی دارد.
3. تفاوت مفهومی «رویه» (procedure) و «تابع» (function) را در برنامه‌نویسی امری توضیح دهد.
4. مدل اجرای فراخوانی تابع در حافظه (رکورد فعال‌سازی، پشته‌ی فراخوانی) را، مستقل از نحو C، ترسیم کند.
5. تابع را در C تعریف و فراخوانی کند: پارامتر، مقدار بازگشتی، پروتوتایپ، دامنه‌ی محلی/سراسری، `static`.
6. بازگشت را با تعریف ریاضیِ استقرایی توجیه کند، حالت پایه و گام بازگشتی را تشخیص دهد، و آن را با مدل پشته‌ی فراخوانی پیوند بزند.
7. نشان دهد که هر حلقه را می‌توان با یک تابع بازگشتی شبیه‌سازی کرد، و درک کند چرا شبیه‌سازیِ معکوس (بازگشت با حلقه) به یک ساختار داده‌ی پشته‌ی صریح نیاز دارد.
8. رابطه‌ی بازگشت با پیمایش DFS درخت فراخوانی را توضیح دهد.

## سطح ۰ — کلی‌ترین تعریف ممکن: تابع به‌عنوان یک ایده، مستقل از هر زمینه

پیش از ریاضیات، پیش از برنامه‌نویسی، حتی پیش از هر فرمول‌بندی رسمی، ایده‌ی «تابع» یک ایده‌ی بسیار ساده و همه‌جا حاضر است:

> **تابع یعنی: قاعده‌ای که هرگاه چیزی مشخص به آن بدهیم، بدون ابهام، چیزی معین به ما پس می‌دهد.**

این ایده مخصوص ریاضیات یا کامپیوتر نیست. یک دستگاه فروش خودکار «تابعی» است که ورودی‌اش (کد کالا + پول) را می‌گیرد و خروجی‌اش (همان کالا) را قطعی و بدون ابهام تحویل می‌دهد. یک دستور آشپزی «تابعی» است که ورودی‌اش (مواد اولیه) را می‌گیرد و طبق مراحلی مشخص، خروجی‌اش (غذا) را تولید می‌کند.

نکته‌ای که در همین سطحِ کاملاً غیررسمی هم می‌توان دید: **برای اینکه یک قاعده واقعاً «تابع» باشد، نتیجه‌اش برای یک ورودیِ ثابت نباید مبهم یا چندگانه باشد.** اگر دستگاه فروش خودکار گاهی یک کالا و گاهی کالای دیگری (بدون تفاوت در ورودی) تحویل دهد، دیگر یک تابع قابل‌اعتماد نیست — این خصلتِ «بدون ابهام»، دقیقاً همان چیزی است که در سطح بعد، ریاضیات آن را دقیق و رسمی می‌کند.

## سطح ۱ — تعریف رسمی در ریاضیات

ریاضیات ایده‌ی سطح قبل را این‌طور دقیق می‌کند:

> یک تابع $f$ از مجموعه‌ی $A$ (**دامنه**، Domain) به مجموعه‌ی $B$ (**برد**، Codomain)، نوشته‌شده $f: A \to B$، قاعده‌ای است که **به هر عضو $A$، دقیقاً یک عضو $B$** نسبت می‌دهد.

کلمه‌ی کلیدی اینجا **«دقیقاً یک»** است. این قسمت از تعریف اختیاری یا سلیقه‌ای نیست — بخشی از خودِ *تعریف* تابع بودن است. اگر قاعده‌ای به یک ورودی، دو خروجیِ متفاوت نسبت دهد، آن قاعده از نظر ریاضی اصلاً «تابع» نامیده نمی‌شود؛ به آن **رابطه** (Relation) می‌گویند — رابطه عمومی‌تر است و تابع فقط حالت خاصی از رابطه است که شرط «تک‌مقداری بودن» را دارد.

### مثال‌های آشنا برای محکم‌کردن تعریف

- $f(x) = x^2$ یک تابع از $\mathbb{R}$ به $\mathbb{R}$ است: هر عدد حقیقی، دقیقاً یک مربع دارد.
- اما «$y$ چنان که $y^2 = x$» یک **تابع نیست** (برای $x=4$، هم $y=2$ و هم $y=-2$ درست‌اند) — این یک رابطه است، نه تابع؛ برای تبدیلش به تابع باید مشخص کنیم کدام ریشه را می‌خواهیم (مثلاً «فقط ریشه‌ی مثبت» → تابع جذر می‌شود).

### یک نکته‌ی مهم که دقیقاً به سؤال بعدی‌مان می‌رسد: خروجی «یک عضو از $B$» می‌تواند خودش یک شیء مرکب باشد

توجه کنید تعریف نمی‌گوید خروجی باید «یک عدد ساده» باشد — می‌گوید باید «یک عضو از $B$» باشد. اگر $B$ را مجموعه‌ای از **زوج‌مرتب‌ها** تعریف کنیم، آنگاه یک تابع می‌تواند در ظاهر «دو عدد» برگرداند، درحالی‌که از نظر ریاضی هنوز دقیقاً **یک** خروجی دارد (یک زوج‌مرتب، که خودش یک عضو واحد از $B$ است). برای مثال:

$$\text{divmod}: \mathbb{Z} \times \mathbb{Z}^{+} \to \mathbb{Z} \times \mathbb{Z}, \qquad \text{divmod}(a, b) = (\lfloor a/b \rfloor,\ a \bmod b)$$

اینجا $\text{divmod}(17, 5) = (3, 2)$ — یک خروجیِ *واحد*، که به‌صورت یک زوجِ (خارج‌قسمت، باقی‌مانده) بسته‌بندی شده. این نکته را در سطح ۶ دوباره می‌بینیم، وقتی به این می‌رسیم که چرا در C معمولاً به‌جای این کار، دو تابع مجزا تعریف می‌کنیم.

## سطح ۲ — از ریاضیات به محاسبه: تابع به‌عنوان چیزی «قابل‌محاسبه»

تعریف ریاضیِ بالا، هیچ ادعایی درباره‌ی *چگونگی* پیدا‌کردن خروجی از روی ورودی نمی‌کند — فقط می‌گوید چنین نسبت‌دهی‌ای وجود دارد. اما وقتی صحبت از **محاسبه** می‌شود، سؤال تازه‌ای مطرح می‌شود:

> آیا یک فرآیندِ گام‌به‌گامِ متناهی (یک **الگوریتم**) وجود دارد که، با گرفتن ورودی، پس از تعداد محدودی گام، به خروجیِ درست برسد؟

این دقیقاً همان مفهوم **الگوریتم** است که در فصل «تفکر الگوریتمی» یاد گرفتیم: توالی، شرط، تکرار. حالا آن مفهوم را به یک نگاه دقیق‌تر پیوند می‌زنیم: **یک تابعِ محاسباتی، چیزی نیست جز یک تابعِ ریاضی که یک الگوریتم برای محاسبه‌اش وجود دارد.** برنامه‌نویسی، عملاً هنرِ *نوشتنِ همین الگوریتم‌ها* به زبانی است که کامپیوتر بفهمد.

(این‌که آیا برای *هر* تابع ریاضی چنین الگوریتمی وجود دارد یا نه، پرسشی عمیق‌تر است — به نظریه‌ی محاسبه‌پذیری و پایان‌ناپذیری مسئله‌ی توقف مربوط می‌شود؛ فعلاً فرض ما این است که با توابعی سروکار داریم که *قابل‌محاسبه*اند، یعنی همان‌هایی که در برنامه‌نویسی روزمره می‌نویسیم.)

## سطح ۳ — رویه (Procedure): مفهومی ساده‌تر و عمومی‌تر از تابع، در برنامه‌نویسیِ امری

زبان‌های **امری** (Imperative، مثل C) برنامه را نه به‌شکل «مجموعه‌ای از تعریف‌های ریاضی»، بلکه به‌شکل **دنباله‌ای از دستورهایی که وضعیت حافظه را قدم‌به‌قدم تغییر می‌دهند** می‌بینند (دقیقاً همان مدلی که در فصل معماری کامپیوتر با آن آشنا شدید: حافظه، و پردازنده‌ای که دستور به دستور آن را تغییر می‌دهد).

در این نگاه، ساده‌ترین و عمومی‌ترین شکلِ «بسته‌بندیِ یک قطعه منطق» را **رویه** (Procedure) می‌نامیم:

> **رویه**، یک دنباله‌ی نام‌گذاری‌شده از دستورها است که می‌توان آن را از جاهای مختلفِ برنامه صدا زد، بدون اینکه لازم باشد آن دنباله را هربار از نو بنویسیم. یک رویه **الزامی به تولید یک خروجیِ مشخص ندارد** — ممکن است فقط کاری انجام دهد (مثلاً چیزی چاپ کند یا مقدار متغیری را عوض کند) بدون اینکه چیزی «برگرداند».

چرا این مفهوم قبل از «تابع» می‌آید؟ چون **عمومی‌تر** است: هر تابع (به معنای ریاضی‌اش) را می‌توان به‌شکل یک رویه پیاده کرد، اما هر رویه‌ای یک تابع ریاضی معادل با خروجیِ مشخص ندارد (مثلاً رویه‌ای که فقط یک پیام چاپ می‌کند، هیچ «خروجی»ای در معنای ریاضی ندارد). انگیزه‌ی اصلیِ رویه هم دقیقاً همان چیزی است که در فصل تفکر الگوریتمی با نام **تجزیه‌ی مسئله و ماژولار بودن** دیدیم: یک منطق تکراری را یک‌بار می‌نویسیم، هر جا لازم شد صدایش می‌زنیم، به‌جای کپی‌پیست‌کردن آن.

## سطح ۴ — از رویه به تابع: افزودنِ «قرارداد یک خروجی»

حالا دو مفهومی که تا اینجا ساختیم را کنار هم می‌گذاریم:

- از **سطح ۱** (ریاضیات): تابع یعنی چیزی که *دقیقاً یک* خروجی مشخص تولید می‌کند.
- از **سطح ۳** (برنامه‌نویسی امری): رویه یعنی دنباله‌ای نام‌گذاری‌شده از دستورها، بدون الزام به تولید خروجی.

**تابع در یک زبان امری مثل C** دقیقاً محل تلاقیِ این دو ایده است:

> تابع = یک رویه + **قراردادی** که می‌گوید این رویه، در پایانِ کارش، *دقیقاً یک* مقدار مشخص تولید می‌کند که می‌توان آن را در جای دیگری از برنامه (مثلاً در یک عبارتِ محاسباتی) مستقیماً استفاده کرد.

این توضیح می‌دهد چرا در C، بعضی رویه‌ها را با کلیدواژه‌ی `void` مشخص می‌کنیم (یعنی: «این یک رویه‌ی خالص است، هیچ قراردادِ خروجی‌ای ندارد») و بقیه یک نوعِ بازگشتی مشخص دارند (یعنی: «این، هم رویه است هم تابع — قول می‌دهد دقیقاً یک مقدار از این نوع برگرداند»).

## سطح ۵ — سمانتیکِ فراخوانی، مستقل از نحوِ C: رکورد فعال‌سازی و پشته‌ی فراخوانی

پیش از دیدنِ نحوِ C، باید بفهمیم فراخوانیِ یک رویه/تابع، **در سطح حافظه و پردازنده** دقیقاً چه اتفاقی است — این مدل به هیچ زبان خاصی وابسته نیست؛ هر زبان امری (C، جاوا، پایتون، ...) از یک نسخه‌ی این مدل استفاده می‌کند.

### مسئله‌ای که باید حل شود

وقتی رویه‌ی $A$، رویه‌ی $B$ را صدا می‌زند، پردازنده باید سه چیز را جایی نگه دارد:

1. **آدرس بازگشت**: بعد از تمام‌شدنِ $B$، اجرا باید دقیقاً به همان نقطه‌ای از $A$ برگردد که فراخوانی از آنجا انجام شده بود.
2. **فضای محلیِ $B$**: پارامترها و متغیرهای محلیِ $B$ باید جایی ذخیره شوند، جدا از فضای $A$ (وگرنه با هم تداخل پیدا می‌کنند).
3. این فضا باید **موقت** باشد: با پایان $B$، دیگر لازم نیست و باید آزاد شود.

### چرا ساختار داده‌ی «پشته» (Stack) دقیقاً همین مسئله را حل می‌کند

فراخوانی‌های تودرتو، طبیعتاً یک الگوی **آخرین‌ورودی، اولین‌خروجی** (LIFO) دارند: اگر $A$، $B$ را صدا بزند و $B$، $C$ را صدا بزند، $C$ باید *اولین* چیزی باشد که تمام می‌شود، بعد $B$، بعد $A$ — دقیقاً برعکسِ ترتیب ورود. این همان رفتاری است که یک **پشته** (Stack) به‌طور طبیعی دارد. به همین دلیل، هر زبان امری، بخشی از حافظه‌ی برنامه را به‌عنوان **پشته‌ی فراخوانی** (Call Stack) کنار می‌گذارد، و هر فراخوانی، یک بلوکِ تازه به‌نام **رکورد فعال‌سازی** یا **قابِ پشته** (Stack Frame) روی آن قرار می‌دهد که شامل همان سه‌چیزِ بالاست.

### تصویر ذهنی: فراخوانی تودرتوی سه رویه

```
زمان t1: A شروع می‌شود                 زمان t2: A، B را صدا می‌زند        زمان t3: B، C را صدا می‌زند
┌─────────────┐                        ┌─────────────┐                    ┌─────────────┐
│   قابِ A     │                        │   قابِ B     │  <- بالای پشته │            قابِ C │  <- بالای پشته
└─────────────┘                        ├─────────────┤                    ├─────────────┤
                                       │   قابِ A     │                    │   قابِ B     │
                                       └─────────────┘                    ├─────────────┤
                                                                          │   قابِ A     │
                                                                          └─────────────┘

زمان t4: C تمام می‌شود، قابش برداشته می‌شود، کنترل به B برمی‌گردد (دقیقاً همان‌جایی که B، C را صدا زده بود)
┌─────────────┐
│   قابِ B     │  <- دوباره بالای پشته
├─────────────┤
│   قابِ A     │
└─────────────┘
```

نکته‌ی کلیدی که در چند خط بعد به آن نیاز داریم: **هر فراخوانی، قابِ خودش را دارد.** یعنی اگر $B$ خودش را دوباره صدا بزند (بازگشت!)، هیچ مشکلی پیش نمی‌آید — هر فراخوانیِ $B$، صرف‌نظر از اینکه چندمین‌بار است، قابِ کاملاً مستقلِ خودش را روی پشته می‌گیرد. این دقیقاً چیزی است که در بخش بازگشت (سطح ۷) به آن برمی‌گردیم.

> 💾 این ناحیه‌ی «پشته» فقط یکی از چند ناحیه‌ی حافظه‌ای است که یک برنامه‌ی در حال اجرا دارد؛ نواحیِ دیگر شامل ناحیه‌ی کدِ ماشین (Text)، داده‌های سراسری/ثابت (Data)، و ناحیه‌ای به نام **Heap** برای حافظه‌ی پویا هستند که چند هفته‌ی دیگر با آن آشنا می‌شوید. همین الان کافی است بدانید: **پشته** برای داده‌های *موقتِ* هر فراخوانی است.

### ۶.۴ پارامتر: `pass-by-value` و اثبات عملیِ مدل رکورد فعال‌سازی

طبق مدل سطح ۵، هر فراخوانی قابِ خودش را دارد. پس وقتی یک متغیر را به‌عنوان پارامتر می‌دهیم، تابع در واقع **کپیِ آن مقدار را در قابِ خودش** می‌گیرد، نه خودِ متغیرِ اصلی را. بیایید این را با چاپِ آدرس‌ها *ببینیم*:


````python
%%writefile pbv.c
#include <stdio.h>

void try_to_modify(int x) {
    printf("   داخل تابع، آدرس کپیِ محلیِ x: %p ، مقدار: %d\n", (void*)&x, x);
    x = 999;
    printf("   داخل تابع، بعد از تغییر: x = %d\n", x);
}

int main(void) {
    int a = 5;
    printf("در main، آدرس a: %p ، مقدار: %d\n", (void*)&a, a);
    try_to_modify(a);
    printf("در main، بعد از فراخوانی: a = %d   <-- تغییر نکرد!\n", a);
    return 0;
}
````

````python
!gcc -Wall -o pbv pbv.c && ./pbv
````

دو آدرسِ چاپ‌شده **کاملاً متفاوت‌اند** — این دقیقاً دلیلِ عملیِ همان چیزی است که در سطح ۵ پیش‌بینی کردیم: `x` داخل `try_to_modify` در یک قابِ کاملاً جدا و مستقل روی پشته نشسته، هیچ ارتباطی با آدرسِ `a` در قابِ `main` ندارد؛ تغییر دادنِ `x` فقط همان کپیِ محلی را عوض می‌کند.

### ۶.۵ دامنه (Scope) و حافظه‌ی `static`

طبق سطح ۵، متغیرهای محلی داخل قابِ پشته زندگی می‌کنند و با پایانِ تابع از بین می‌روند. اما گاهی می‌خواهیم یک متغیر محلی، **بین فراخوانی‌های مختلف مقدارش را حفظ کند** — یعنی جایی *خارج از* پشته زندگی کند. کلیدواژه‌ی `static` دقیقاً همین را می‌گوید: «این متغیر را در ناحیه‌ی داده‌ی سراسری (Data، نه پشته) بگذار، فقط یک‌بار مقداردهی اولیه کن، و اجازه بده بین فراخوانی‌ها زنده بماند.»


````python
%%writefile staticdemo.c
#include <stdio.h>

int next_id(void) {
    static int counter = 0;   // فقط بار اول اجرا می‌شود؛ خارج از پشته زندگی می‌کند
    counter++;
    return counter;
}

int main(void) {
    printf("%d\n", next_id());
    printf("%d\n", next_id());
    printf("%d\n", next_id());
    return 0;
}
````

````python
!gcc -Wall -o staticdemo staticdemo.c && ./staticdemo
````

اگر `static` را حذف کنید، `counter` هر بار از نو روی پشته ساخته و با مقدار ۰ آغاز می‌شود — همیشه `1` چاپ می‌شد. تفاوت رفتار، مستقیماً از تفاوتِ *محلِ حافظه* (پشته در برابر ناحیه‌ی داده‌ی سراسری) می‌آید، نه یک قاعده‌ی دلبخواهی زبان.

## سطح ۶.۵ (تکمیلی) — چرا در C هم هر تابع دقیقاً یک خروجی دارد؟

حالا که هم تعریف ریاضی (سطح ۱) و هم نحو C (سطح ۶) را داریم، می‌توانیم دقیق‌تر از قبل به این سؤال برگردیم: طبق سطح ۱، تابعِ $\text{divmod}$ کاملاً معتبر است و «دو عدد» برمی‌گرداند — چون آن دو عدد را در قالبِ *یک* زوج‌مرتب بسته‌بندی کرده‌ایم. اما زبانِ C (برخلاف بعضی زبان‌های دیگر) نحوِ ساده‌ای برای «برگرداندنِ یک زوج‌مرتب» به شما نمی‌دهد — نوعِ بازگشتیِ یک تابع در C باید یک نوعِ ساده (`int`, `double`, …) باشد.

راه‌حلِ رایج در C دقیقاً همان کاری است که ریاضی هم اجازه می‌دهد: **به‌جای یک تابعِ زوج‌خروجی، دو تابعِ مستقل و تک‌خروجی تعریف می‌کنیم** — هرکدام یک تابعِ کاملاً معتبر طبق سطح ۱، فقط روی مسئله‌ی یک‌بخشی از همان محاسبه:


````python
%%writefile divmod.c
#include <stdio.h>

int quotient(int a, int b) {
    return a / b;
}

int modulo(int a, int b) {
    return a % b;
}

int main(void) {
    int a = 17, b = 5;
    printf("%d / %d = خارج‌قسمت %d ، باقی‌مانده %d\n",
           a, b, quotient(a, b), modulo(a, b));
    return 0;
}
````

````python
!gcc -Wall -o divmod divmod.c && ./divmod
````

دو تابعِ `quotient` و `modulo`، هرکدام دقیقاً یک ورودی-زوج می‌گیرند و دقیقاً یک عدد برمی‌گردانند — کاملاً منطبق با تعریفِ سطح ۱. عملگرهای داخلیِ C یعنی `/` و `%` هم دقیقاً همین الگو را در سطحِ خودِ زبان پیاده کرده‌اند: به‌جای یک عملگر که زوج برگرداند، دو عملگرِ جداگانه داریم.

> ➡️ **پیوند رو به جلو:** چند هفته‌ی دیگر، وقتی با اشاره‌گر آشنا شوید، راهِ دیگری هم می‌بینید: تابعی که به‌جای «برگرداندن» یک مقدار، آدرسِ محلی را می‌گیرد و مستقیماً *در آن محل می‌نویسد* (اصطلاحاً «پارامتر خروجی»). آن روش امکان می‌دهد یک تابع، چند مقدار را هم‌زمان از طریق چند پارامتر خروجی بنویسد؛ اما حتی آنجا هم، نوعِ بازگشتیِ رسمیِ خودِ تابع (اگر داشته باشد) همچنان دقیقاً یکی است — معمولاً برای گزارشِ موفقیت/شکستِ عملیات. فعلاً، تفکیک به دو تابعِ مستقل (مثل بالا) ساده‌ترین و رایج‌ترین راه‌حل است.

## سطح ۷ — بازگشت: از تعریفِ استقراییِ ریاضی تا کد

### ۷.۱ سمانتیک، پیش از هر کد: تعریف‌های بازگشتی/استقرایی در ریاضی

خیلی از مفاهیمِ ریاضی نه با یک فرمولِ مستقیم، بلکه با **تعریف استقرایی** (Inductive Definition) ساخته می‌شوند: یک یا چند **حالتِ پایه** (Base Case) به‌طور مستقیم مشخص می‌شوند، و بقیه‌ی موارد بر اساسِ مواردِ *کوچک‌تر* تعریف می‌شوند. مثال کلاسیک، فاکتوریل:

$$
n! =
\begin{cases}
1 & n = 0 \quad \text{(حالت پایه)} \\
n \times (n-1)! & n > 0 \quad \text{(گامِ استقرایی)}
\end{cases}
$$

نکته‌ی مهم: این تعریف **خودارجاع** دارد ($n!$ بر اساسِ $(n-1)!$ تعریف شده)، اما مشکلی ندارد چون هر بار به سمتِ حالتِ پایه (که مستقیم و بدون خودارجاع تعریف شده) **پیش می‌رود**؛ اگر این پیشروی وجود نداشت (مثلاً اگر می‌گفتیم $n! = n \times n!$)، تعریف ناقص و بی‌معنا می‌شد — دقیقاً همان چیزی که در ۷.۴ به‌عنوان یک باگِ واقعی می‌بینیم.

### ۷.۲ بازگشت در محاسبه: دقیقاً همان مدلِ پشته‌ی سطح ۵، یک لایه عمیق‌تر

طبق سطح ۵، هر فراخوانیِ تابع، قابِ مستقلِ خودش را روی پشته می‌گیرد. **بازگشت چیزی نیست جز اینکه یک تابع، خودش را صدا بزند** — و چون هر فراخوانی (حتی فراخوانیِ خودِ تابع) قابِ خودش را می‌گیرد، هیچ تناقضی پیش نمی‌آید: هر لایه از بازگشت، نسخه‌ی کاملاً مستقلِ خودش از پارامترها و متغیرهای محلی را دارد.


````python
%%writefile factorial.c
#include <stdio.h>

long long factorial(int n) {
    if (n == 0) return 1;             // حالت پایه — دقیقاً همان مورد مستقیمِ تعریف ریاضی
    return n * factorial(n - 1);      // گام بازگشتی — دقیقاً همان گامِ استقرایی
}

int main(void) {
    for (int i = 0; i <= 6; i++) {
        printf("%d! = %lld\n", i, factorial(i));
    }
    return 0;
}
````

````python
!gcc -Wall -o factorial factorial.c && ./factorial
````

### ۷.۴ نتیجه‌ی طبیعیِ مدل: سرریز پشته (Stack Overflow)

اگر گامِ بازگشتی هرگز به حالتِ پایه *نزدیک* نشود، قاب‌ها همچنان روی پشته انباشته می‌شوند تا حافظه‌ی اختصاصی‌شده به پشته (که محدود است) تمام شود — این خطا را **سرریز پشته** می‌نامند. این یک اشتباه واقعی و رایج است: فراموش‌کردنِ کاهشِ آرگومان به‌سمتِ حالتِ پایه:


````python
%%writefile buggy_factorial.c
#include <stdio.h>

long long buggy_factorial(int n) {
    if (n == 0) return 1;
    return n * buggy_factorial(n);   // باگ: باید n-1 باشد، نه n !
}

int main(void) {
    printf("در حال محاسبه...\n");
    fflush(stdout);
    printf("%lld\n", buggy_factorial(5));
    return 0;
}
````

````python
!gcc -Wall -O0 -o buggy_factorial buggy_factorial.c
!timeout 5 ./buggy_factorial ; echo "کد خروج: $?   (139 یعنی Segmentation Fault به‌خاطر سرریز پشته)"
````

همان‌طور که مدل سطح ۵ پیش‌بینی می‌کند: هر فراخوانی یک قابِ تازه اضافه می‌کند و چون `n` هیچ‌وقت به `0` نمی‌رسد، این افزودن هرگز متوقف نمی‌شود تا حافظه‌ی پشته تمام شده و برنامه با خطای دسترسیِ حافظه (Segmentation Fault) کرش می‌کند.

### ۷.۵ درختِ فراخوانی، و چرا بازگشت طبیعتاً یک **پیمایشِ عمقی (DFS)** است

فاکتوریل یک بازگشتِ «خطی» بود — هر فراخوانی دقیقاً یک فراخوانیِ دیگر تولید می‌کرد. اما در حالتِ کلی، هر فراخوانی می‌تواند **چند** فراخوانیِ دیگر تولید کند؛ در این حالت، ساختارِ کاملِ فراخوانی‌ها دیگر یک زنجیره‌ی ساده نیست، بلکه یک **درخت** است — به آن **درختِ فراخوانی** یا **درختِ بازگشت** (Recursion Tree/Call Tree) می‌گویند. مثالِ کلاسیک: دنباله‌ی فیبوناچی، که هر جمله (به‌جز دو جمله‌ی اول) از **دو** جمله‌ی قبلی ساخته می‌شود:

$$\text{fib}(n) = \begin{cases} n & n \le 1 \\ \text{fib}(n-1) + \text{fib}(n-2) & n > 1 \end{cases}$$

بیایید همین را پیاده کنیم، اما این‌بار هر ورود و خروج از تابع را با تورفتگیِ متناسب با عمقِ فراخوانی چاپ می‌کنیم — این کار، **دقیقاً همان چیزی است که پردازنده در پشته انجام می‌دهد را روی صفحه‌ی نمایش می‌آورد**:


````python
%%writefile fibtrace.c
#include <stdio.h>

int call_count = 0;

void print_indent(int depth) {
    for (int i = 0; i < depth; i++) printf("│  ");
}

long fib(int n, int depth) {
    call_count++;
    print_indent(depth);
    printf("┌─ ورود  fib(%d)\n", n);

    long result;
    if (n <= 1) {
        result = n;
    } else {
        long a = fib(n - 1, depth + 1);   // اول کاملاً می‌رویم پایین (عمق) تا ته این شاخه
        long b = fib(n - 2, depth + 1);   // بعد شاخه‌ی بعدی را کاملاً می‌رویم پایین
        result = a + b;
    }

    print_indent(depth);
    printf("└─ خروج  fib(%d) = %ld\n", n, result);
    return result;
}

int main(void) {
    long result = fib(4, 0);
    printf("\nfib(4) = %ld   (تعداد کل فراخوانی‌ها: %d)\n", result, call_count);
    return 0;
}
````

````python
!gcc -Wall -o fibtrace fibtrace.c && ./fibtrace
````

نگاه کنید به ترتیبِ چاپ: `fib(4)` قبل از رفتن سراغِ شاخه‌ی `fib(2)` سمتِ راستش، **کاملاً و تا انتها** وارد شاخه‌ی چپش (`fib(3)`) می‌شود و تا ته آن پایین می‌رود، قبل از این‌که برگردد و شاخه‌ی بعدی را امتحان کند. این *دقیقاً* تعریفِ **پیمایشِ عمقی (Depth-First Search / DFS)** است: اول تا جای ممکن عمیق برو، بعد برگرد و شاخه‌ی بعدی را امتحان کن.

> 🔑 **این یک شباهتِ اتفاقی نیست.** بازگشت، DFS را «مجانی» و به‌طور خودکار پیاده می‌کند، دقیقاً به این دلیل که پشته‌ی فراخوانی (سطح ۵) خودش یک ساختارِ **LIFO** است — و پیاده‌سازیِ استانداردِ DFS هم همیشه با یک پشته (چه صریح، چه ضمنیِ زبان) انجام می‌شود. اگر می‌خواستیم به‌جای DFS، پیمایشِ سطحی (Breadth-First Search / BFS) روی همین درختِ فراخوانی انجام دهیم (یعنی اول همه‌ی فراخوانی‌های عمقِ ۱، بعد همه‌ی عمقِ ۲، …)، بازگشتِ ساده دیگر کافی نیست — BFS به یک **صف** (FIFO) نیاز دارد، نه پشته؛ پیاده‌سازیِ آن با حلقه و یک صفِ صریح ممکن است، نه با بازگشتِ ساده‌ی بالا.

## سطح ۸ — هم‌ارزیِ محاسباتیِ حلقه و بازگشت

حالا با دو ابزار (حلقه از فصلِ قبل، و بازگشت از همین جلسه) یک سؤالِ عمیق‌تر مطرح می‌کنیم: **آیا این دو ابزار، از نظرِ آنچه می‌توانند محاسبه کنند، واقعاً متفاوت‌اند، یا فقط دو نحوِ متفاوت برای یک قدرتِ یکسان‌اند؟**

پاسخِ شناخته‌شده در نظریه‌ی محاسبه این است: **حلقه (همراه با متغیرهای قابل‌تغییر) و بازگشت، از نظرِ محاسباتی هم‌ارزند** — هرچیزی که با یکی قابلِ‌محاسبه است، با دیگری هم قابلِ‌محاسبه است (این نتیجه با نام‌هایی مثل قضیه‌ی برنامه‌ی ساخت‌یافته (Structured Program Theorem) و نظریه‌ی توابعِ بازگشتی در منطقِ ریاضی شناخته می‌شود). بیایید این ادعا را از هر دو جهت با کد بسنجیم.

### ۸.۱ از حلقه به بازگشت: هر حلقه را می‌توان یک تابعِ بازگشتی نوشت

ایده: بدنه‌ی حلقه را در تابع می‌گذاریم، و به‌جای این‌که شمارنده را در یک متغیر «به‌روزرسانی» کنیم، آن را **به‌عنوان پارامترِ فراخوانیِ بعدی** پاس می‌دهیم.


````python
%%writefile sumcompare.c
#include <stdio.h>

int sum_loop(int n) {
    int total = 0;
    for (int i = 1; i <= n; i++) {
        total += i;
    }
    return total;
}

int sum_recursive(int n) {
    if (n == 0) return 0;                    // حالت پایه: مطابقِ حلقه‌ای که صفر بار اجرا می‌شود
    return n + sum_recursive(n - 1);         // گام بازگشتی: مطابقِ یک دورِ حلقه
}

int main(void) {
    for (int n = 0; n <= 5; n++) {
        printf("n=%d:   sum_loop=%d   sum_recursive=%d\n", n, sum_loop(n), sum_recursive(n));
    }
    return 0;
}
````

### ۸.۲ از بازگشت به حلقه: چرا این جهت به یک «پشته‌ی صریح» نیاز دارد

جهتِ عکس هم درست است، اما پیچیده‌تر: طبقِ سطح ۵، بازگشت به این دلیل کار می‌کند که *زبان*، خودش یک پشته‌ی فراخوانی برایمان مدیریت می‌کند (آدرسِ بازگشت، متغیرهای محلی، ...). اگر بخواهیم همان تابعِ بازگشتی را **فقط با حلقه** (بدون فراخوانیِ توابعِ تودرتو) بازنویسی کنیم، باید کارِ پشته‌ی فراخوانی را **خودمان دستی** با یک ساختارِ داده‌ی پشته انجام دهیم:

> **الگوریتمِ کلی:** یک پشته‌ی صریح (که هفته‌ی بعد، پس از یادگیریِ آرایه، می‌توانیم واقعاً پیاده‌اش کنیم) می‌سازیم. به‌جای فراخوانیِ واقعیِ تابع، وضعیتِ لازم (مثلاً مقدارِ `n`) را روی این پشته «هُل می‌دهیم» (Push). در یک حلقه، مرتب از بالای پشته یک وضعیت برمی‌داریم (Pop)، اگر به حالتِ پایه رسیده بود پاسخ را مستقیم حساب می‌کنیم، وگرنه وضعیت‌های جدید (مطابقِ فراخوانی‌های بازگشتیِ اصلی) را دوباره روی پشته هُل می‌دهیم. این حلقه دقیقاً همان‌کاری را می‌کند که خودِ زبان در سطح ۵ برایمان انجام می‌داد، فقط این‌بار با کدِ خودمان.

توجه کنید این الگوریتم، **دقیقاً همان ترتیبِ DFS** بخشِ ۷.۵ را بازتولید می‌کند — چون از یک پشته (LIFO) استفاده می‌کند. اگر به‌جای پشته از یک **صف** (FIFO) استفاده کنیم، همان الگوریتم به یک پیمایشِ **BFS** روی درختِ فراخوانی تبدیل می‌شود — ترتیبِ محاسبه‌ی متفاوتی از بازگشتِ معمولی، که برای خودِ فاکتوریل یا فیبوناچیِ ساده لزومی ندارد ولی در مسائلی مثل «کوتاه‌ترین مسیر» (که در دروسِ بعدی می‌بینید) حیاتی است.

> ⏭️ چون پیاده‌سازیِ واقعیِ این پشته‌ی صریح به آرایه (یا لیستِ پیوندی، که بعدتر می‌بینید) نیاز دارد، پیاده‌سازیِ کاملِ آن را به بعد از فصلِ آرایه موکول می‌کنیم — همین‌جا فقط *ایده* و *چرایی*اش را دقیق دانستیم.

## 📚 پیشنهادهای مطالعاتی تکمیلی (خارج از هسته‌ی الزامیِ درس)

### ۱. برنامه‌نویسیِ تابعی (Functional Programming): وقتی بازگشت تنها ابزارِ تکرار است

در این جلسه، حلقه و بازگشت را دو ابزارِ *هم‌ارز* دیدیم که هر دو در C در دسترس‌اند. اما خانواده‌ای از زبان‌های برنامه‌نویسی به‌نامِ **زبان‌های تابعی** (Functional، مثل Haskell، یا زیرمجموعه‌ای از پایتون/جاوااسکریپت) با فلسفه‌ای متفاوت طراحی شده‌اند: در این زبان‌ها، **متغیرها معمولاً غیرقابل‌تغییر (Immutable)** هستند — یعنی وقتی مقداری را به یک نام نسبت دادید، دیگر نمی‌توانید آن نام را به مقدارِ دیگری تغییر دهید.

چرا این مهم است؟ به یاد بیاورید که حلقه‌ی `for` در C برای کار کردن به یک متغیرِ *قابل‌تغییر* (مثلِ شمارنده‌ی `i` یا انباره‌ی `total`) نیاز دارد. اگر تغییرِ متغیر مجاز نباشد، حلقه به همین شکلِ آشنا دیگر امکان‌پذیر نیست — و این دقیقاً همان چیزی است که زبان‌های تابعی را وادار می‌کند **تقریباً برای هر تکراری از بازگشت استفاده کنند** (دقیقاً همان `sum_recursive` بخشِ ۸.۱، نه `sum_loop`). به‌طور نمادین (نه کدِ واقعیِ اجراشدنی)، تعریفِ مجموعِ ۱ تا $n$ در سبکِ تابعی چیزی شبیهِ این است:

```
sum(n) = 0                     اگر n = 0
sum(n) = n + sum(n - 1)        در غیر این صورت
```

که دقیقاً همان $\text{sum\_recursive}$ ماست — فقط این‌بار **تنها** راهِ ممکن، نه یک انتخابِ جایگزین.

### ۲. بهینه‌سازیِ فراخوانیِ دنباله‌ای (Tail-Call Optimization)

مشکلِ آشکار: اگر بازگشت تنها ابزارِ تکرار باشد، آیا هر تکرارِ طولانی باعثِ سرریزِ پشته نمی‌شود (طبقِ بخشِ ۷.۴)؟ بسیاری از زبان‌های تابعی (و برخی کامپایلرهای C در حالتِ بهینه‌سازی) یک ترفند به‌نامِ **بهینه‌سازیِ فراخوانیِ دنباله‌ای** دارند: اگر آخرین کاری که یک تابع انجام می‌دهد *دقیقاً* فراخوانیِ خودش باشد (نه این‌که نتیجه‌ی آن فراخوانی را در محاسبه‌ی دیگری استفاده کند)، کامپایلر می‌تواند قابِ فعلی را *بازاستفاده* کند به‌جای اضافه‌کردنِ قابِ تازه — و بازگشت را عملاً به یک حلقه تبدیل می‌کند، بدونِ رشدِ پشته. توجه کنید `factorial` ما (بخشِ ۷.۲) این شکل *نیست*، چون بعد از فراخوانیِ بازگشتی، هنوز باید در نتیجه ضرب کند؛ `sum_recursive` هم دقیقاً به همین دلیل تابعِ دنباله‌ای نیست. نوشتنِ نسخه‌ی دنباله‌ایِ این توابع (با یک پارامترِ اضافه برای انباشتِ نتیجه‌ی جزئی) موضوعِ جالبی برای کاوشِ شخصی است.

### ۳. توابعی که بازگشتی‌اند اما «ساده» نیستند: تابعِ آکرمان

همه‌ی توابعِ بازگشتی به همان راحتیِ فاکتوریل قابلِ تبدیل به حلقه نیستند. **تابعِ آکرمان** یک مثالِ معروف در نظریه‌ی محاسبه است: کاملاً محاسبه‌پذیر و همیشه پایان‌پذیر است، اما آن‌قدر سریع رشد می‌کند و آن‌قدر عمیق بازگشت می‌کند که هیچ فرمولِ بسته یا حلقه‌ی ساده‌ای معادلِ آن (بدونِ خودِ بازگشتِ تودرتو) شناخته‌نشده — نمونه‌ای عالی برای دیدنِ اینکه بازگشت می‌تواند از حلقه‌های «ساده» (به‌نامِ بازگشتِ ابتدایی/Primitive Recursive) قوی‌تر باشد.

### ۴. وقتی درختِ بازگشت زیرمسئله‌های تکراری دارد: مموایزیشن

به نمودارِ `fib(4)` در بخشِ ۷.۵ دوباره نگاه کنید: `fib(2)` **دوبار** کامل محاسبه شد (یک‌بار زیرِ شاخه‌ی `fib(3)`، یک‌بار مستقیم). برای `fib` عددهای بزرگ‌تر، این همپوشانی به‌شکلِ نمایی بدتر می‌شود. تکنیکِ **مموایزیشن** (Memoization) پیشنهاد می‌دهد: نتیجه‌ی هر فراخوانی را (مثلاً در یک آرایه، که هفته‌ی بعد می‌بینید) ذخیره کنیم؛ پیش از محاسبه‌ی دوباره‌ی یک زیرمسئله، اول چک کنیم آیا قبلاً محاسبه شده — این ایده، پایه‌ی حوزه‌ای به‌نامِ **برنامه‌نویسیِ پویا** (Dynamic Programming) است که در دروسِ الگوریتم عمیق‌تر می‌بینید.

### 🧩 یادداشت برای بازاستفاده از این قالب در مبحث بعدی («آرایه»)
برای مبحثِ بعدی، همین جهتِ کلی‌به‌جزء را حفظ کنید: شروع از سؤالِ کاملاً کلی («چطور چند مقدار مرتبط را با هم مدیریت کنیم؟»)، سپس مدلِ آن در حافظه (بلوکِ پیوسته، طبقِ فصلِ معماری) **پیش از** نحوِ `int arr[N]`، و در پایان، بازگشت به وعده‌ی همین نوت‌بوک: پیاده‌سازیِ واقعیِ پشته‌ی صریح برای شبیه‌سازیِ بازگشت با حلقه (بخشِ ۸.۲).
